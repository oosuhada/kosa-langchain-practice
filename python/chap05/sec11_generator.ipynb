```python
generator = (
    v for v in range(1, 1000001)
)  # 이 range 안에서 생성할 것이다 라는 의미. 아직 생성된 것은 아니다. 제너레이터는 필요할 때마다 값을 생성한다.
print(
    type(generator)
)  # <class 'generator'> - ( )로 감싸면 제너레이터 표현식이 된다. tuple 아님 주의
# 만약 (v for v in range(1, 1000001))가 아니라 [v for v in range(1, 1000001) ]로 작성하면 리스트 컴프리헨션이 된다.
# 리스트 컴프리헨션은 모든 값을 한 번에 생성하여 메모리에 저장해버려서 메모리를 많이 사용한다.

# for가 요청할때 비로소 generator가 값을 생성한다. range(1, 1000001)에서 1부터 1000000까지의 숫자를 하나씩 생성한다.
total = 0  # sum() 내장 함수를 가리지 않는 습관!
for v in generator:
    total += v
print(total)  # 500000500000
# 코드는 비슷하지만 제너레이터를 사용하면 메모리를 훨씬 적게 사용한다.
# 제너레이터는 한 번에 하나의 값만 생성하기 때문에 메모리를 효율적으로 사용할 수 있다.
```

```python
from typing import Generator

# 외부 처리값을 받아서 재가공해서 리턴하는 제너레이터 생성 함수 정의
# 첫번째 list: 제너레이터가 제공하는 값의 타입
# 두번째 list: 외부에서 가공처리한 값을 받는 타입
# 세번째 str: 제너레이터가 두번째에서 받은 값을 재가공해서 반환하는 값의 타입


def create_generator() -> Generator[list, list, str]:
    # 외부에 줄 list값 생성
    source: list = [1, 2, 3, 4, 5]
    # 외부에 리스트를 제공
    print("1) 외부에 리스트를 제공")
    print("2) 외부에서 처리가 될때까지 대기")
    result: list = yield source
    # 재가공 처리
    print("4) 외부 처리 결과를 재가공 처리")
    total = 0  # sum() 내장 함수를 가리지 않는 습관!
    for v in result:
        total += v
    final_result = f"제너레이터에서 재가공한 결과값: {total}"
    print("5) 재가공 결과를 반환")
    return final_result
```

```python
import time

# 외부에서 제너레이터 사용하기

# 제너레이터 생성
generator = create_generator()

# 제너레이터로부터 값 가져오기
source: list = next(generator)  # 제너레이터로부터 source 리스트를 받아온다.

# 제너레이터로부터 가져온 값을 처리하기
result: list = [value * 10 for value in source]

# 처리된 결과를 다시 제너레이터로 보내기
try:
    time.sleep(5)
    print("3) 외부 처리 완료")
    generator.send(
        result
    )  # 제너레이터로 result 리스트를 보내준다. 제너레이터는 이 값을 받아서 재가공 처리한다.
    # send(result)는 멈춰 있던 yield 표현식 자리에 result를 전달하고 이어서 실행한다.
except StopIteration as e:
    print(e.value)  # 제너레이터가 종료되었으므로 제너레이터를 다시 사용할 수 없다.
```

```python
import random
import time
from typing import Generator

# 실행 순서 확인용 번호
step = 1


def print_step(message: str) -> None:
    global step
    print(f"{step}) {message}")
    step += 1


# 초기 온도값 생성
value = random.randint(0, 30)
print_step(f"초기 온도값 설정: {value}도")


# 온도 센서 데이터를 계속 제공하는 제너레이터 생성 함수 정의
# 첫번째 dict: 제너레이터가 외부에 제공하는 값의 타입
# 두번째 None: 외부에서 send()로 보내주는 값 없음
# 세번째 None: 제너레이터가 return으로 반환하는 값 없음
def temperature_sensor(value: int) -> Generator[dict, None, None]:
    print_step("제너레이터 내부 실행 시작")

    while True:
        # 외부에 제공할 센서 데이터 생성
        sensor_data: dict = {
            "sensor": "temperature",
            "value": value,
        }

        # 외부에 센서 데이터 제공
        print_step("외부에 현재 온도 데이터를 제공")
        print_step("외부에서 다음 next()가 호출될 때까지 대기")
        yield sensor_data

        # 외부에서 next()가 다시 호출되면 yield 다음 줄부터 재개
        print_step("외부에서 next() 재호출 완료")
        print_step("제너레이터 내부에서 온도값 갱신 처리")

        value += random.randint(-1, 1)

        print_step(f"갱신된 온도값: {value}도")
        print_step("다음 센서 데이터 제공 전 5초 대기")
        time.sleep(5)
        print_step("5초 대기 완료")
```

```python
# 외부에서 제너레이터 사용하기

# 외부에서 제너레이터 객체 생성
temperature_sensor_gen = temperature_sensor(value)
print_step("제너레이터 객체 생성 완료")
print_step("아직 제너레이터 함수 내부는 실행되지 않음")

# 제너레이터로부터 첫 번째 센서 데이터 가져오기
print_step("외부에서 첫 번째 next() 호출")
sensor_data: dict = next(temperature_sensor_gen)
print_step(f"외부에서 센서 데이터 수신: {sensor_data}")

# 제너레이터로부터 두 번째 센서 데이터 가져오기
print_step("외부에서 두 번째 next() 호출")
sensor_data: dict = next(temperature_sensor_gen)
print_step(f"외부에서 센서 데이터 수신: {sensor_data}")

# for문으로 제너레이터 값 계속 가져오기
# for문은 내부적으로 next()를 반복 호출한다.
# time.sleep(5)가 있어서 5초마다 센서 데이터가 출력된다.
# KeyboardInterrupt, 즉 Ctrl + C로 중단 가능하다.
print_step("for문으로 센서 데이터 자동 수신 시작")

for sensor_data in temperature_sensor_gen:
    print_step(f"외부에서 센서 데이터 수신: {sensor_data}")
```
