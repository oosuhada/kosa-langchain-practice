```python
var = 1
print(var)  # 1 → 전역 (네임스페이스에 선언된) 변수 var 읽기


def fun1():
    global var  # 함수 안에서 전역 변수 var를 직접 사용하겠다는 선언
    print(var)  # 전역 변수 var 읽기 → 1
    var = 2  # 전역 변수 var를 2로 변경 (전역 변수에 영향을 줌)
    # global var이 없다면 var = 2는 함수 내부 (네임스페이스에 선언된) 지역 변수
    # global var 선언이 없으면 var = 2는 지역 변수 var을 새로 만드는 것 → 전역 변수 var에는 영향 없음
    # 주의: global var 선언은 함수 맨 위에 해야 함 → 중간에 선언하면 SyntaxError
    print(var)  # 2


fun1()
print(var)  # 2 → fun1() 호출 후 전역 변수가 바뀌어 있음
```

```python
var = 1


def fun1():
    global var  # 전역 변수 var를 직접 사용하겠다는 선언
    var = var + 1  # 전역 변수 var를 읽어서 +1 후 전역 변수 var에 재할당
    print(var)  # 2


fun1()
print(var)  # 2 → fun1() 호출 후 전역 변수가 바뀌어 있음


def fun2():
    global var  # 전역 변수 var를 직접 사용하겠다는 선언
    var += 1  # var = var + 1 과 동일 → 전역 변수 var에 1을 더해서 재할당
    print(var)  # 3


fun2()
print(var)  # 3 → fun2() 호출 후 전역 변수가 바뀌어 있음


# global 선언 없이 var = var + 1 또는 var += 1 을 사용하면?
def fun3():
    var += 1  # ❌ UnboundLocalError
    # 파이썬이 var += 1 을 보고 var를 지역변수로 인식
    # 그런데 지역변수 var는 선언된 적 없으므로 읽기 시도 시 에러 발생
    # var = var + 1 도 동일하게 UnboundLocalError 발생


# global 선언 없이 var = 2 만 하면?
def fun4():
    var = var + 1  # ❌ UnboundLocalError → fun3()과 동일한 이유
    # var = 2 는 지역변수를 새로 만드는 것 → 전역 변수 var에는 영향 없음
    # var = var + 1 은 지역변수 var를 읽으려 하지만 아직 선언 안됨 → 에러
```
