```python
# Python은 객체 참조를 전달한다. 불변 객체는 재할당해도 원본이 바뀌지 않고, 가변 객체는 내부를 수정하면 원본에 영향이 보인다.
# 불변 객체: 객체 내부 수정 불가 - int, float, str, bool, tuple 등
# 가변 객체: 객체 내부 수정 가능 - list, dict, set 등


def fun1(param: int) -> None:
    param = 10
    # param은 var1의 복사본 → var1과 다른 메모리 주소
    # param = 10 은 복사본만 바꾸는 것 → var1에 영향 없음


def fun2(param: list) -> None:
    param[0] = 10
    # param은 var2의 주소를 가리킴 → var2와 같은 메모리 주소
    # param[0] = 10 은 같은 주소의 값을 직접 수정 → var2도 같이 바뀜
```

```python
# immutable: Call By Value (값에 의한 호출)
var1 = 5
fun1(var1)
print("var1:", var1)  # var1: 5
# 불변객체 var1의 값은 fun1 함수 내부에서 변경되지 않고 그대로 유지된다. (immutable)
# int, str, tuple 같은 불변 타입은 함수에 "값만 복사해서" 전달
# 즉 함수 안에서 아무리 바꿔도 원본은 절대 안 바뀜
```

```python
# mutable: Call By Reference (참조에 의한 호출)
var2 = [1, 2, 3]
fun2(var2)
print("var2:", var2)  # var2: [10, 2, 3]
# 가변객체 var2의 값은 fun2 함수 내부에서 변경되어 fun2 함수 외부에서도 변경된 값이 유지된다. (mutable)
# list, dict, set 같은 가변 타입은 함수에 "주소(참조)를 전달"
# 즉 함수 안에서 수정하면 원본도 같이 바뀜
```

```python
var2 = [1, 2, 3]
fun2(var2)
print("var2:", var2)
```
