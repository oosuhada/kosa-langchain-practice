```python
# 함수 정의
def score_avg(scores: list[int]) -> float:
    total = sum(scores)
    count = len(scores)
    average = total / count
    return average


def score_avg2(scores: list[int]) -> float:
    sum: int = 0

    for i in range(len(scores)):
        sum += scores[i]

    avg: float = sum / len(scores)
    return avg


# 함수 호출
scores: list[int] = [90, 80, 70, 60, 50]
result: float = score_avg(scores)
print(result)
```

```python
def start_program() -> None:
    print("전원을 시작합니다.")
    print("로봇을 시작합니다.")


def move_program() -> None:
    print("로봇이 움직입니다.")
    print("로봇이 저장된 위치에 도착합니다.")


def stop_program() -> None:
    print("로봇이 멈춥니다.")
    print("로봇이 전원을 끕니다.")


# exit은 대화형 환경에서 쓰이는 이름이라 수업 예제에서는 피하는 게 좋음.
def exit_program() -> None:
    print("프로그램을 종료합니다.")


while True:
    print("명령어 입력", end="")
    command: str = input()
    if command == "start" or command == "1":
        start()
    elif command == "move" or command == "2":
        move()
    elif command == "stop" or command == "3":
        stop()
    elif command == "exit" or command == "4":
        print("프로그램을 종료합니다.")
        break
    else:
        print("잘못된 명령어입니다. 다시 입력해주세요.")
```
