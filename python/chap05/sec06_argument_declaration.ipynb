#### - 가변 길이 매개변수


```python
# *(아스타)의 의미 = 여러개의 값이 주어질 수 있다. 가변 길이 변수다
# 함수 정의에서 매개변수 이름 앞에 *를 붙이면, 그 매개변수는 가변 길이 인자를 받는 매개변수가 된다.
# *를 통해 "몇개가 들어올지 모르니까 다 받아서 튜플로 묶어줘" 라는 의미가 된다.
def fun(*values: int) -> None:
    print(type(values), values)
    for i in range(len(values)):
        print(f"values[{i}]:", values[i])


# fun(1, 2, 3) 하면 파이썬이 자동으로 내부적으로 values는 (1, 2, 3)이라는 튜플로 묶어서 전달한다.

fun(1, 2, 3)  # <class 'tuple'> (1, 2, 3)
```

```python
def fun2(a, b, *c, d, e=7) -> None:
    print(a, b, c, d, e)


fun2(1, 2, d=6)  # a=1, b=2, c=() (빈 튜플), d=6, e=7 (기본값)
fun2(1, 2, 3, 4, 5, 6, d=6)  # a=1, b=2, c=(3, 4, 5), d=6, e=7 (기본값)
fun2(
    1, 2, 3, 4, 5, e=100, d=6
)  # a=1, b=2, c=(3, 4, 5), d=6, e=100 (기본값 대신 명시적으로 전달)
# fun2(1, 2, 3, 4, 5, d=6, e=100) 순서는 상관없다. 키워드 인자는 이름으로 전달하기 때문에 순서가 중요하지 않다.
```

#### - 키워드 매개 변수


```python
# *(아스타) 단독으로 사용하는 경우 그 뒤에 선언된 매개변수들이 모두 키워드 매개변수가 된다
# 가변 길이 매개변수 또한 키워드 매개변수에 속한다.


def fun1(*a, b, c):  # *a에 가변길이로 값을 받겠다는 의미, b와 c는 키워드 전용
    print(a, b, c)


fun1(1, 2, 3, b=4, c=5)  # a=(1, 2, 3), b=4, c=5
fun1(b=4, c=5)  # a=(), b=4, c=5 → *a는 아무것도 안 받으면 빈 튜플


def fun2(*, b, c):  # * 뒤에 나오는 매개변수들은 모두 키워드 매개변수라는 의미
    print(b, c)


fun2(b=1, c=2)  # 1 2
fun2(c=2, b=1)  # 1 2 → 키워드 인자는 순서 상관없음
# fun2(1, 2)      # TypeError: 위치 인자로 전달 불가
# fun2(1, c=2)    # TypeError: 첫 번째 인자도 키워드로 전달해야 함
# fun2(b=1, 2)    # SyntaxError: 키워드 인자 뒤에 위치 인자 불가
```

```python
def fun(a, b, *, c, d="Python"):
    print(a, b, c, d)


fun(1, 2, c=3)  # a=1, b=2, c=3, d='Python' (기본값)
fun(1, 2, c=3, d="Java")  # a=1, b=2, c=3, d='Java' (기본값 대신 명시적으로 전달)
# fun (1, 2, 3) 이라고 사용하면 TypeError가 발생하므로 세 번째 인자 c는 키워드로 전달해야 함
```

```python
# *와 /의 차이: *는 키워드 전용 인자, /는 위치 전용 인자
# /는 *와 달리 fun(1, 2, 3, 4)가 가능하다.
# / 앞에 선언된 매개변수들은 반드시 위치 인자로만 전달해야 한다.
# / 뒤에 선언된 매개변수들은 위치 인자와 키워드 인자 모두 가능하다.
# * 뒤에 선언된 매개변수들은 반드시 키워드 인자로만 전달해야 한다.


def fun(a, b, /, c, d):  # a, b는 위치 전용 / c, d는 위치or키워드 둘 다 가능
    print(a, b, c, d)


fun(1, 2, 3, 4)  # ✅ a=1, b=2, c=3, d=4
fun(1, 2, c=3, d=4)  # ✅ a=1, b=2, c=3, d=4
fun(1, 2, d=4, c=3)  # ✅ a=1, b=2, c=3, d=4
# fun(a=1, b=2, c=3, d=4)  # ❌ TypeError: a, b는 키워드로 전달 불가 (/ 앞이라서)
# fun(1, 2, c=3, 4)         # ❌ SyntaxError: 키워드 인자 뒤에 위치 인자 불가
# fun(1, 2, 3, d=4)         # ✅ a=1, b=2, c=3, d=4 → c는 위치로, d는 키워드로 가능


# * 와 / 동시에 사용하는 경우
# a, b → 위치 전용 (/ 앞)
# c, d → 위치or키워드 둘 다 가능 (/ 뒤, * 앞)
# e, f → 키워드 전용 (* 뒤)
def fun2(a, b, /, c, d, *, e, f):
    print(a, b, c, d, e, f)


fun2(1, 2, 3, 4, e=5, f=6)  # ✅ a=1, b=2, c=3, d=4, e=5, f=6
fun2(1, 2, c=3, d=4, e=5, f=6)  # ✅ a=1, b=2, c=3, d=4, e=5, f=6
# fun2(a=1, b=2, c=3, d=4, e=5, f=6)  # ❌ TypeError: a, b는 키워드 전달 불가
# fun2(1, 2, 3, 4, 5, 6)              # ❌ TypeError: e, f는 위치 전달 불가


# * vs / 핵심 정리
# def fun(*, a, b)  → a, b 반드시 키워드로만   → fun(a=1, b=2) ✅  fun(1, 2) ❌
# def fun(a, b, /)  → a, b 반드시 위치로만     → fun(1, 2) ✅      fun(a=1, b=2) ❌
```

#### - 가변 길이 키워드 매개 변수


```python
# **kwargs: 키워드 가변 길이 매개변수
# *args와 달리 키워드 인자(key=value)를 받아서 딕셔너리로 묶어준다

# 기본 사용법
def fun(**a):  # 키워드 인자를 모두 딕셔너리로 묶어서 받겠다는 의미
    print(type(a), a)
    for key in a:
        print(
            f"a[{key}]:", a[key]
        )  # 방식1. key가 없으면 KeyError 발생 (지금은 모든 key가 존재하므로 에러 안남)
        print(
            f"a.get({key}):", a.get(key)
        )  # 방식2. key값이 없으면 None 반환 (에러 안남)


fun(x=1, y=2, z=3)  # <class 'dict'> {'x': 1, 'y': 2, 'z': 3}
fun(name="홍길동", age=30)  # <class 'dict'> {'name': '홍길동', 'age': 30}
# fun(1, 2, 3)  # ❌ TypeError: 위치 인자로 전달 불가


# 위치 인자 + 키워드 가변 길이
def fun2(a, **b):  # a는 위치 인자, 나머지 키워드 인자는 모두 b에 딕셔너리로 묶임
    print(a, b)


fun2(1, x=2, y=3)  # 1 {'x': 2, 'y': 3}
fun2(1)  # 1 {} → **b는 아무것도 안 받으면 빈 딕셔너리
# fun2(x=2, y=3)          # ❌ TypeError: a는 반드시 위치 인자로 전달해야 함


# 가변 길이 위치 인자 + 가변 길이 키워드 인자
def fun3(*a, **b):  # *a는 위치 인자를 튜플로, **b는 키워드 인자를 딕셔너리로
    print(a, b)


fun3(1, 2, 3, x=4, y=5)  # (1, 2, 3) {'x': 4, 'y': 5}
fun3(1, 2, 3)  # (1, 2, 3) {} → **b는 빈 딕셔너리
fun3(x=4, y=5)  # () {'x': 4, 'y': 5} → *a는 빈 튜플


# 키워드 전용 인자 + 키워드 가변 길이
def fun4(*, a, b, **c):  # a, b는 키워드 전용 / 나머지 키워드 인자는 c에 딕셔너리로
    print(a, b, c)


fun4(a=1, b=2)  # 1 2 {}
fun4(a=1, b=2, x=3, y=4)  # 1 2 {'x': 3, 'y': 4}
# fun4(1, 2)                # ❌ TypeError: a, b는 키워드로만 전달 가능


# * vs ** 핵심 정리
# *args  → 위치 인자 여러 개를 튜플로      → fun(1, 2, 3)
# **kwargs → 키워드 인자 여러 개를 딕셔너리로 → fun(x=1, y=2)
```

```python
def fun(a, **b):
    print(
        a, b
    )  # 위치 고정 인자 a는 위치로 전달, 나머지 키워드 인자는 b에 딕셔너리로 묶임


fun(1, x=2, y=3)  # 1 {'x': 2, 'y': 3}
```

#### - 딕셔너리 언패킹


```python
# 딕셔너리 언패킹
person = {"name": "Gabriel", "age": 34}

# 딕셔너리를 그냥 언패킹하면 키만 나옴
name, age = person
print(name, age)  # name age

# 값을 언패킹하려면 .values() 사용
name, age = person.values()
print(name, age)  # Gabriel 34

# 키와 값을 함께 언패킹하려면 .items() 사용
for key, value in person.items():
    print(key, value)  # name Gabriel / age 34


# 리스트 언패킹 → 값만 담으면 순서대로 언패킹
person = ["Gabriel", 34]  # 리스트는 키 없이 값만
name, age = person
print(name, age)  # Gabriel 34


# 딕셔너리는 키가 반드시 문자열 또는 불변타입이어야 함
# 값(value)은 어떤 타입이든 가능
person = {"name": "Gabriel", "age": 34}  # 키는 str, 값은 str/int 혼합 가능
name, age = person.values()
print(name, age)  # Gabriel 34
```

```python
def fun(**args):
    print(args)


person = {"name": "Gabriel", "age": 34}
# **args는 가변길이 키워드 매개변수의 의미였고, **person은 딕셔너리를 키워드 인자로 풀어서 전달하는 언패킹의 의미가 된다.
# **person은 딕셔너리 person의 key-value 쌍을 키워드 인자로 풀어서 fun 함수에 전달한다.
fun(
    **person
)  # {'name': 'Gabriel', 'age': 34} → **person은 딕셔너리를 키워드 인자로 풀어서 전달
```

```python
def fun(*, mid, mname, mage):
    print(mid, mname, mage)


member = {"mid": "user123", "mname": "Gabriel", "mage": 34}
fun(
    **member
)  # mid, mname, mage" 키워드 인자로 전달 → **member는 딕셔너리를 키워드 인자로 풀어서 전달
fun(mid="user123", mname="Gabriel", mage=34)  # **member 대신 키워드 인자로 직접 전달
```
