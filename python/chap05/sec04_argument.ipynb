```python
import datetime

boards = []


# 게시물 추가 함수
# 기본값을 가지는 매개변수는 기본값이 없는 매개변수 뒤에 위치해야 한다.
def add_board(
    title: str,
    content: str,
    writer: str,
    date: str = datetime.datetime.now().strftime("%Y-%m-%d"),  # 기본값
    hitcount: int = 0,  # 기본값
) -> None:
    board = {
        "title": title,
        "content": content,
        "writer": writer,
        "date": date,
        "hitcount": hitcount,
    }
    boards.append(board)


# 함수 실행. date, hitcount 생략 가능
add_board("오늘은 비가 온다", "오늘 날씨가 좋지 않네요.", "홍길동")
add_board("오늘은 맑다", "오늘 날씨가 좋네요.", "홍길동")

# date 직접 지정도 가능
add_board("오늘은 흐리다", "내용", "홍길동", date="2026-01-01")

print(boards)
```
