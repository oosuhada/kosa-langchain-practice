```python
# 내장 함수와 동일한 이름으로는 전역 변수나 함수 정의하지 않기. 정의하는 순간 내장 함수가 가려져서 사용할 수 없게 됨.

list()  # cities = list() 와 cities = [] 는 같은 의미.
tuple()  # season = tuple(["봄", "여름", "가을", "겨울"]) 와 season = ("봄", "여름", "가을", "겨울") 는 같은 의미.
dict()  # member = dict(name="Gabriel", age=34) 와 member = {"name": "Gabriel", "age": 34} 는 같은 의미.
set()  # unique_numbers = set([1, 2, 3, 4, 5]) 와 unique_numbers = {1, 2, 3, 4, 5} 는 같은 의미.

# list(), tuple(), dict(), set()은 내장 타입을 함수 호출 형태로 호출해 새 객체를 만든다.
# 객체 생성 코드. 생성자 호출 -> 함수가 아니라 클래스. 생성자를 호출해서 객체를 만드는것.
# 엄격히 말하자면 함수는 아니지만, 함수 호출과 같은 형태로 보이기 때문에 함수라고 부르기도 함.
```

```python
print(len([1, 2, 3]))  # 3
print(max(1, 10, 5))  # 10
print(min(1, 10, 5))  # 1
```
