```python
# import mypack 패키지 자체는 import 대상이 아님. 패키지 내부의 모듈을 import 해야 함
import mypack.module1
import mypack.module2
import mypack.module3

mypack.module1.fun()
mypack.module2.fun()
mypack.module3.fun()
```

```python
# mypack 패키지에서 필요한 모듈만 import한다.
from mypack import module1, module2, module3

module1.fun()
module2.fun()
module3.fun()
```

```python
# __init__.py 파일에 __all__ = ["module1", "module2", "module3"] 설정 후에 *로 import 할 수 있다.
from mypack import *

module1.fun()
module2.fun()
module3.fun()
```

```python
from mypack.module1 import fun

fun()
```

```python
# 모듈로 부터 모든 정의된 멤버(변수, 함수, 클래스)를  *로 가져올 수 있음
from mypack.module1 import *

fun()
```

```python
# 패키지와 계층들이 복잡하게 구성되는 경우에 __init__.py 파일에 from .module1 import fun 선언 후에 fun()을 호출할 수 있음
from mypack import fun

fun()
```

```python
from mypack.subpack import module4

module4.fun()
```

```python
from mypack.subpack.module4 import fun

fun()
```
