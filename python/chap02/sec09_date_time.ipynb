```python
from datetime import datetime

now = datetime.now()

print(now, type(now))  # 현재 날짜와 시간, 그리고 그 타입을 출력한다. datetime.now() 함수는 현재 날짜와 시간을 나타내는 datetime 객체를 반환한다. 출력 결과는 현재 날짜와 시간에 대한 정보와 함께 <class 'datetime.datetime'>이라는 타입 정보가 포함된다.
print(str(now.year)+"년")
print(str(now.month)+"월")
print(str(now.day)+"일")
print(str(now.hour)+"시")
print(str(now.minute)+"분")
print(f"현재 datetime: {now.year}년 {now.month}월 {now.day}일 {now.hour}시 {now.minute}분")  # f-string을 사용하여 현재 날짜와 시간을 형식화하여 출력한다. f-string은 문자열 내에서 중괄호 {}를 사용하여 변수나 표현식을 삽입할 수 있는 기능을 제공한다. 출력 결과는 현재 날짜와 시간이 "년 월 일 시 분" 형식으로 표시된다.
```

```python
# 간단하게 출력하는법
print(now.strftime("%Y.%m.%d"))
print(now.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
# datetime 생성자를 이용해서 지정한 날짜와 시간을 나타내는 datetime 객체를 생성할 수 있다. datetime(year, month, day, hour, minute, second) 형식으로 사용한다.

target_datetime = datetime(2050, 12, 25, 0, 0, 0)  # 특정 날짜와 시간을 나타내는 datetime 객체를 생성한다. d-day
print(target_datetime)  # 생성된 datetime 객체를 출력한다. 출력 결과는 2050년 12월 25일 00시 00분 00초를 나타내는 datetime 객체의 정보가 포함된다.
```
