```python
import re

num1 = 3
num2 = 5
result = num1 + num2
print(result)

# java에서는 개행(줄바꿈)을 세미콜론(;)으로 구분하지만, 파이썬에서는 개행 자체가 명령어의 끝을 나타낸다.
# result = num1 
#           + num2;
# 따라서 세미콜론은 필요하지 않으며, 사용하지 않는 것이 일반적이다.
# 파이썬에서 줄바꿈을 사용하고 싶다면 괄호를 사용한다
result = (num1 
            + num2)
```

```python
str1 = "Python "
str2 = "Programming"
result_str = str1 + str2
print(result_str)
# .py 파이썬 스크립트 파일에서는 반드시 print() 함수를 사용하여 결과를 출력해야 한다.
# 그러나 인터프리터나 Jupyter Notebook에서는 마지막 줄의 표현식이 자동으로 출력된다. 
result_str2 = "Python " + "Programming"
print(result_str2)

result_str3 = "Python " "Programming"
print(result_str3)

result_str4 = ("Python "
               "Programming")
print(result_str4)
```

```python
name = "Gabriel"
age = 34
result_eng = name + " is " + str(age) + " years old."
print(result_eng)
result_kor = name + "님의 나이는 " + str(age) + "세 입니다."
print(result_kor)

# 파이썬 3.6 이상에서는 f-string을 사용하여 문자열을 더 간결하고 읽기 쉽게 작성할 수 있다.
# f-string은 문자열 앞에 f 또는 F를 붙이고, 중괄호 {} 안에 변수나 표현식을 넣어 문자열을 포맷팅하는 방법이다.
# f stands for "formatted string literal"의 약자이다.
result_fstring = f"{name} is {age} years old."
print(result_fstring)
result_fstring_kor = f"{name}님의 나이는 {age}세 입니다."
print(result_fstring_kor)
result_fstring_kor = (f"{name}님의 나이는 "
                            f"{age}세 입니다.")
print(result_fstring_kor)
```
