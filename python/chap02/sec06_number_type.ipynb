```python
result = 10 / 4
print(result)  # 2.5
# Java에서는 정수 나눗셈이 정수 결과를 반환하지만, 파이썬에서는 항상 부동 소수점 결과를 반환한다.
```

```python
result = 10**4
print(result)  # 10000
# Java에서는 Math.pow(10, 4)와 같이 제곱을 계산하지만, 파이썬에서는 ** 연산자를 사용하여 제곱을 계산할 수 있다
```

```python
result = 10 // 4
print(result)  # 2
# // 연산자는 정수 나눗셈을 수행하여 결과를 정수로 반환한다. 이는 Java의 정수 나눗셈과 유사하다.
# 정수끼리 계산하면 int, float가 섞이면 float가 될 수 있다.
```

```python
result = 10 % 4
print(result)  # 2
# % 연산자는 나머지 연산자로, Java와 동일하게 작동한다. 10을 4로 나눈 나머지는 2이다.
```

```python
result = 10
result += 5
print(result)  # 15
# Java에서는 result = result + 5와 같이 작성하지만, 파이썬에서는 result += 5와 같이 축약된 연산자를 사용할 수 있다.
# 이는 result에 5를 더한 후 다시 result에 할당하는 것을 의미한다.
```

```python
result = 10
result *= 5
print(result)  # 50
# Java에서는 result = result * 5와 같이 작성하지만, 파이썬에서는 result *= 5와 같이 축약된 연산자를 사용할 수 있다.
# 이는 result에 5를 곱한 후 다시 result에 할당하는 것을 의미한다.
```

```python
var1 = 100
var2 = 100
print(var1 == var2)  # True. 값 비교(동등 비교)
print(var1 is var2)  # True. 번지 비교. 작은 정수는 캐싱되어 동일한 객체를 참조
```
