#### - 비교 연산

```python
num1 = 5
num2 = 10

print(num1 == num2) #동등 비교. 객체 안의 값 비교.
print(num1 is num2) #번지 비교. 객체의 주소값 비교. num1과 num2는 서로 다른 객체를 참조하기 때문에 False.
print(num1 != num2) #동등 비교의 반대. 객체 안의 값 비교.
print(num1 is not num2) #번지 비교의 반대. 객체의 주소값 비교. num1과 num2는 서로 다른 객체를 참조하기 때문에 True.
```

#### - 논리 연산

```python
num1 = 5
num2 = 10

print(num1 > 3 and num2 < 10) #논리 연산자 and. num1이 3보다 크고(num1 > 3) num2가 10보다 작은(num2 < 10) 경우에만 True를 반환한다. num1은 3보다 크지만(num1 > 3) num2는 10보다 작지 않기 때문에(num2 < 10) 전체 표현식은 False.
print(num1 > 3 or num2 < 10) #논리 연산자 or. num1이 3보다 크거나(num1 > 3) num2가 10보다 작은(num2 < 10) 경우에 True를 반환한다. num1은 3보다 크기 때문에(num1 > 3) 전체 표현식은 True.
print(not(num1 > 3 and num2 < 10)) #논리 연산자 not. num1이 3보다 크고(num1 > 3) num2가 10보다 작은(num2 < 10) 경우에 True를 반환하는 표현식의 결과를 반전
print(not(num1 > 3 or num2 < 10)) #논리 연산자 not. num1이 3보다 크거나(num1 > 3) num2가 10보다 작은(num2 < 10) 경우에 True를 반환하는 표현식의 결과를 반전
# 파이썬에서 not 연산자는 다른 언어에서 ! 연산자와 유사한 역할을 한다. not 연산자는 논리값을 반전시키는 역할을 한다. 예를 들어, not True는 False가 되고, not False는 True가 된다.
```

```python
str1 = "abc"
str2 = "Abc"
print(str1 == str2)  # False. 문자열이 대소문자를 구분
print(str1 is str2)  # False. str1과 str2는 서로 다른 객체를 참조한다.
```
