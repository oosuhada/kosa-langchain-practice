#### - 문자열 인덱싱과 슬라이싱


```python
text = "파이썬 프로그래밍"

# 문자열 인덱싱1
print(text[0])  # 파. 문자열의 첫 번째 문자에 접근한다. 인덱스는 0부터 시작한다.
print(text[1])  # 이. 문자열의 두 번째 문자에 접근한다.
print(text[2])  # 썬. 문자열의 세 번째 문자에 접근한다.
print(text[-1])  # 밍. 문자열의 마지막 문자에 접근한다.
print(text[-2])  # 래. 문자열의 끝에서 두 번째 문자에 접근
print(text[-3])  # 그. 문자열의 끝에서 세 번째 문자에 접근한다.

# 문자열 슬라이싱
# 슬라이싱 문법: [시작인덱스:끝인덱스:스텝 크기(선택적)]
print(
    text[0:3]
)  # 파이썬. 문자열의 첫 번째 문자부터 세 번째 문자까지 슬라이싱한다. 인덱스 3은 포함되지 않는다.
print(
    text[4:11]
)  # 프로그래밍. 문자열의 다섯 번째 문자부터 열한 번째 문자까지 슬라이싱한다. 인덱스 11은 포함되지 않는다.
print(
    text[4:11:2]
)  # 프그밍. 문자열의 다섯 번째 문자부터 열한 번째 문자까지 슬라이싱한다. 인덱스 11은 포함되지 않는다. 2는 스텝 크기를 나타낸다.
print(text[4:])  # 프로그래밍. 문자열의 다섯번째 문자부터 마지막 문자까지 슬라이싱한다.
print(
    text[-5:]
)  # 프로그래밍. 문자열의 끝에서 다섯 번째 문자부터 마지막 문자까지 슬라이싱한다.
print(text[:3])  # 파이썬. 문자열의 처음부터 세 번째 문자까지 슬라이싱한다.
```

```python
numbers = "123456789"
print(
    numbers[::2]
)  # 13579. 문자열의 처음부터 끝까지 슬라이싱한다. 2는 스텝 크기를 나타낸다. 즉, 문자열에서 짝수 인덱스에 해당하는 문자만 추출한다.
print(
    numbers[1::2]
)  # 2468. 문자열의 처음부터 끝까지 슬라이싱한다. 2는 스텝 크기를 나타낸다. 즉, 문자열에서 홀수 인덱스에 해당하는 문자만 추출한다.
```

```python
# 주민번호에서 생년월일과 뒷자리 7자리를 따로 출력하시오
# 주민번호 뒷자리에 따라 남자인지 여자인지 출력하시오

ssn = "900101-1234567"
print(
    ssn[:6]
)  # 900101. 주민번호에서 생년월일을 슬라이싱한다. 처음부터 인덱스 6까지 슬라이싱한다.
print(
    ssn[7:]
)  # 1234567. 주민번호에서 뒷자리 7자리를 슬라이싱한다. 인덱스 7부터 마지막까지 슬라이싱한다.

# 혹은 변수를 사용하면
first = ssn[
    :6
]  # 주민번호에서 생년월일을 슬라이싱한다. 처음부터 인덱스 6까지 슬라이싱한다.
last = ssn[
    7:
]  # 주민번호에서 뒷자리 7자리를 슬라이싱한다. 인덱스 7부터 마지막까지 슬라이싱한다.
print(first)  # 900101
print(last)  # 1234567

# 혹은 split() 메서드를 사용하여 주민번호를 생년월일과 뒷자리로 분리할 수도 있다.
first = ssn.split(
    "-"
)[
    0
]  # 주민번호를 하이픈(-)을 기준으로 분리한다. split() 메서드는 문자열을 지정된 구분자로 나누어 리스트로 반환한다.
last = ssn.split(
    "-"
)[
    1
]  # 주민번호를 하이픈(-)을 기준으로 분리한다. split() 메서드는 문자열을 지정된 구분자로 나누어 리스트로 반환한다.
print(first)  # 900101
print(last)  # 1234567

gender_id = ssn[
    7
]  # 주민번호에서 성별을 나타내는 숫자를 추출한다. 인덱스 7은 주민번호 뒷자리의 첫 번째 숫자이다.
if gender_id in ["1", "3"]:  # 성별이 1 또는 3이면 남자이다.
    print("남자")
elif gender_id in ["2", "4"]:  # 성별이 2 또는 4이면 여자이다.
    print("여자")
else:
    print("알 수 없는 성별")

# 혹은 if-else 문으로 성별을 판별할 수도 있다.
if gender_id == "1" or gender_id == "3":
    print("남자")
else:
    print("여자")

# 혹은 if-elif-else 문 대신 삼항 연산자를 사용하여 성별을 판별할 수도 있다.
gender = (
    "남자"
    if gender_id in ["1", "3"]
    else "여자"
    if gender_id in ["2", "4"]
    else "알 수 없는 성별"
)
print(
    gender
)  # 삼항 연산자를 사용하여 성별을 판별한다. JavaScript/Java에서 조건식 ? 참일 때 값 : 거짓일 때 값의 형태로 작성했었다.
# Python의 삼항 표현식은 참일_때_값 if 조건식 else 거짓일_때_값 형태다.
# 중첩된 삼항 연산자를 사용하여 여러 조건을 처리할 수 있다.


# 혹은 if-elif-else 문 대신 딕셔너리를 사용하여 성별을 판별할 수도 있다.
gender_dict = {"1": "남자", "2": "여자", "3": "남자", "4": "여자"}
print(
    gender_dict.get(gender_id, "알 수 없는 성별")
)  # 딕셔너리에서 성별을 판별한다. get() 메서드는 딕셔너리에서 키에 해당하는 값을 반환한다. 키가 존재하지 않으면 두 번째 인자로 지정된 기본값을 반환한다.
```

#### - 문자열 비교


```python
var1 = "Python"
var2 = "python"
var3 = "Python"
print(
    var1 == var2
)  # False. 값 비교(동등 비교). 대소문자를 구분하기 때문에, var1과 var2는 다른 문자열로 간주된다
print(
    var1 == var3
)  # True. 값 비교(동등 비교). 대소문자를 구분하기 때문에, var1과 var3는 같은 문자열로 간주된다.
print(var1 is var2)  # False. 번지 비교. var1과 var2는 서로 다른 객체를 참조한다.
print(
    var1 is var3
)  #문자열 리터럴은 인터닝 때문에 같은 객체를 참조할 수 있다. 값 비교에는 is가 아니라 ==를 사용한다.
```

```python
var1 = "Python"
var2 = "Python" + "3"
var3 = "Python" + "3"
print(
    var2 == var3
)  # True. 값 비교(동등 비교). var2와 var3는 같은 문자열 값을 가지므로 True.
print(var2 is var3)  # True. 번지(객체) 비교.
# "Python" + "3" 처럼 컴파일 시점에 확정되는 리터럴 연산은
# 파이썬이 컴파일 타임에 "Python3"으로 최적화(폴딩)하고,
# 문자열 인터닝(interning)을 통해 동일한 객체를 재사용한다.
# 따라서 var2와 var3는 동일한 객체를 참조한다.
# ※ "캐싱"은 정확한 표현이 아님. 정확히는 컴파일 타임 상수 폴딩 + 인터닝.

var1 = "Python"
var2 = var1 + "3"
var3 = var1 + "3"
print(
    var2 == var3
)  # True. 값 비교(동등 비교). var2와 var3는 같은 문자열 값을 가지므로 True.
print(var2 is var3)  # False. 번지(객체) 비교.
# var1 + "3" 은 런타임에 실행되는 변수 연산이므로,
# 호출할 때마다 새로운 문자열 객체가 생성된다.
# 파이썬의 인터닝은 컴파일 타임에 확정된 리터럴에만 적용되며,
# 런타임 변수 연산 결과에는 적용되지 않는다.
# 따라서 var2와 var3는 같은 값("Python3")을 가지지만
# 서로 다른 메모리 주소에 존재하는 별개의 객체를 참조한다.

# 그러므로 값 비교를 할때는 == 연산자를 사용하고, 객체 비교를 할 때는 is 연산자를 사용하는 것이 좋다.
```

#### - 문자열 포함 여부


```python
var = "Python" in "This is Python programming language."
print(
    var
)  # True. "Python"이라는 문자열이 "This is Python programming language." 문자열 안에 포함되어 있기 때문에 True. in 연산자는 왼쪽 피연산자가 오른쪽 피연산자에 포함되어 있는지를 검사하여 True 또는 False를 반환한다

var = "Java" in "This is Python programming language."
print(
    var
)  # False. "Java"라는 문자열이 "This is Python programming language." 문자열 안에 포함

var = "Python" not in "This is Python programming language."
print(
    var
)  # False. "Python"이라는 문자열이 "This is Python programming language." 문자열 안에 포함

var = "Java" not in "This is Python programming language."
print(
    var
)  # True. "Java"라는 문자열이 "This is Python programming language." 문자열 안에 포함되어 있지 않음

var = "python" in "This is Python programming language."
print(
    var
)  # False. 대소문자를 구분하기 때문에, "python"이라는 문자열이 "This is Python programming language." 문자열 안에 포함되어 있지 않음

# 문자열 비교는 대소문자를 구분하기 때문에, 대소문자를 구분하지 않고 비교하기 위해서는 문자열을 모두 소문자 또는 모두 대문자로 변환한 후 비교한다
var = "python" in "This is Python programming language.".lower()
print(
    var
)  # True. "This is Python programming language." 문자열을 모두 소문자로 변환한 후 "python"이라는 문자열이 포함되어 있는지 검사하기 때문에 True.

# 함수는 독립적이고 메소드는 객체에 종속되어서 객체 안에서 작동하는 함수이다.
# .lower() 메서드는 .을 통해서 문자열 객체에 접근
# len()은 문자열의 길이를 반환하는 독립적인 함수이다
```

#### - 문자열 검사


```python
str1 = "abc123456def"
print(
    str1.isalpha()
)  # False. 문자열이 알파벳으로만 구성되어 있는지 검사한다. str1에는 숫자도 포함되어 있기 때문에 False.
print(
    str1.isdigit()
)  # False. 문자열이 숫자로만 구성되어 있는지 검사한다. str1에는 알파벳도 포함되어 있기 때문에 False.
print(
    str1.isalnum()
)  # True. 문자열이 알파벳과 숫자로만 구성되어 있는지 검사한다. str1에는 알파벳과 숫자만 포함되어 있기 때문에 True.
```

```python
str2 = "3"
str3 = "3.5"
print(
    str2.isdigit()
)  # True. 문자열이 숫자로만 구성되어 있는지 검사한다. str2는 숫자로만 구성되어 있기 때문에 True.
print(
    str3.isdigit()
)  # False. 문자열이 숫자로만 구성되어 있는지 검사한다. str3에는 소수점(.)이 포함되어 있기 때문에 False.

print(
    int(str2)
)  # 3. 문자열을 정수로 변환한다. str2는 숫자로만 구성되어 있기 때문에 정수로 변환할 수 있다.
# print(int(str3))  # ValueError: invalid literal for int() with base 10: '3.5'. 문자열을 정수로 변환하려고 시도하지만, str3에는 소수점(.)이 포함되어 있기 때문에 정수로 변환할 수 없다. int() 함수는 문자열이 정수 형태일 때만 변환이 가능하다.
print(
    float(str3)
)  # 3.5. 문자열을 실수로 변환한다. str3는 소수점이 포함된 숫자로 구성되어 있기 때문에 실수로 변환할 수 있다. float() 함수는 문자열이 실수 형태일 때 변환이 가능하다.
```

#### - 문자열 공백 제거


```python
text = input()
print(
    text.strip()
)  # 입력한 문자열의 양쪽 끝에 있는 공백을 제거한다. strip() 메서드는 문자열의 양쪽 끝에 있는 공백을 제거하는 데 사용된다. 입력한 문자열의 양쪽 끝에 공백이 있을 경우, 출력 결과에서는 공백이 제거된 문자열이 출력된다.
print(text.lstrip())  # 입력한 문자열의 왼쪽 끝에 있는 공백을 제거한다. 잘 사용안함
print(text.rstrip())  # 입력한 문자열의 오른쪽 끝에 있는 공백을 제거한다. 잘 사용안함
```

```python
str = input("문자열을 입력하세요: ")
str2 = str.strip()
print(str2)
print(
    len(str2)
)  # 입력한 문자열의 길이를 출력한다. strip() 메서드로 양쪽 끝의 공백이 제거된 문자열의 길이를 계산하여 출력한다.
```

#### - 문자열 찾기


```python
str = "123456-3890123"
print(
    str.index("-")
)  # 6. 문자열에서 하이픈(-)의 인덱스를 반환한다. index() 메서드는 문자열에서 지정된 값이 처음으로 나타나는 위치의 인덱스를 반환한다. 하이픈(-)은 인덱스 6에 위치한다.
print(
    str.find("-")
)  # 6. 문자열에서 하이픈(-)의 인덱스를 반환한다. find() 메서드는 문자열에서 지정된 값이 처음으로 나타나는 위치의 인덱스를 반환한다. 하이픈(-)은 인덱스 6에 위치한다. index() 메서드와 달리, find() 메서드는 찾는 값이 없을 때 -1을 반환한다.
print(
    str.find("-", 7, 9)
)  # -1. 인덱스 7~8 범위에서 하이픈을 검색. 해당 범위에 없으므로 -1 반환.

seperator = str.find("-")
if seperator != -1:
    part1 = str[
        :seperator
    ]  # 하이픈(-)을 기준으로 문자열을 슬라이싱한다. 처음부터 하이픈의 인덱스까지 슬라이싱한다.
    part2 = str[
        seperator + 1 :
    ]  # 하이픈(-)을 기준으로 문자열을 슬라이싱한다. 하이픈의 인덱스 + 1부터 마지막까지 슬라이싱한다.
    print(part1)  # 123456
    print(part2)  # 3890123
else:
    print("하이픈(-)이 문자열에 포함되어 있지 않습니다.")
```

#### - 토큰 분리


```python
tel_num = "010-1234-5678"
seperator1 = tel_num.find("-")
seperator2 = tel_num.find("-", seperator1 + 1)
if seperator1 != -1 and seperator2 != -1:
    part1 = tel_num[
        :seperator1
    ]  # 하이픈(-)을 기준으로 문자열을 슬라이싱한다. 처음부터 첫 번째 하이픈의 인덱스까지 슬라이싱한다.
    part2 = tel_num[
        seperator1 + 1 : seperator2
    ]  # 하이픈(-)을 기준으로 문자열을 슬라이싱한다. 첫 번째 하이픈의 인덱스 + 1부터 두 번째 하이픈의 인덱스까지 슬라이싱한다.
    part3 = tel_num[
        seperator2 + 1 :
    ]  # 하이픈(-)을 기준으로 문자열을 슬라이싱한다. 두 번째 하이픈의 인덱스 + 1부터 마지막까지 슬라이싱한다.
    print(part1)  # 010
    print(part2)  # 1234
    print(part3)  # 5678
else:
    print("하이픈(-)이 문자열에 충분히 포함되어 있지 않습니다.")


parts = tel_num.split(
    "-"
)  # 문자열을 하이픈(-)을 기준으로 분리한다. split() 메서드는 문자열을 지정된 구분자로 나누어 리스트로 반환한다.
if len(parts) == 3:
    part1, part2, part3 = (
        parts  # 리스트 언패킹을 사용하여 분리된 문자열을 각각의 변수에 할당한다.
    )
    print(part1)  # 010
    print(part2)  # 1234
    print(part3)  # 5678
else:
    print("하이픈(-)이 문자열에 충분히 포함되어 있지 않습니다.")

parts2 = tel_num.split("-")
print(
    parts2
)  # ['010', '1234', '5678']. 문자열을 하이픈(-)을 기준으로 분리하여 리스트로 반환한다. split() 메서드는 문자열을 지정된 구분자로 나누어 리스트로 반환한다.
print(
    parts2[0]
)  # 010. 문자열을 하이픈(-)을 기준으로 분리하여 리스트로 반환한 후, 리스트의 첫 번째 요소에 접근한다.
print(
    parts2[1]
)  # 1234. 문자열을 하이픈(-)을 기준으로 분리하여 리스트로 반환한 후, 리스트의 두 번째 요소에 접근한다.
print(
    parts2[2]
)  # 5678. 문자열을 하이픈(-)을 기준으로 분리하여 리스트로 반환한 후, 리스트의 세 번째 요소에 접근한다.

telephone_number = input("전화번호를 입력하세요 (예: 010-1234-5678): ")
parts = telephone_number.split(
    "-"
)  # 문자열을 하이픈(-)을 기준으로 분리한다. split() 메서드는 문자열을 지정된 구분자로 나누어 리스트로 반환한다.
if len(parts) == 3:
    part1, part2, part3 = (
        parts  # 리스트 언패킹을 사용하여 분리된 문자열을 각각의 변수에 할당한다.
    )
    print(
        "입력한 전화번호:", parts
    )  # 입력한 전화번호: 010-1234-5678. 사용자가 입력한 전화번호를 그대로 출력한다.
    print("국번:", part1)  # 국번: 010
    print("중간 번호:", part2)  # 중간 번호: 1234
    print("끝 번호:", part3)  # 끝 번호: 5678
else:
    print(
        "입력한 전화번호 형식이 올바르지 않습니다. 하이픈(-)으로 구분된 세 부분으로 입력해주세요."
    )
```
