```python
# 파이썬은 모두 객체로 취급하기 때문에, 모든 데이터는 객체로 간주된다. 따라서 모든 데이터는 클래스의 인스턴스이다.
v1 = "Hello"
print(type(v1))  # <class 'str'>
```

```python
v2 = 10
print(type(v2))  # <class 'int'>
```

```python
v3 = 3.14
print(type(v3))  # <class 'float'>
# 파이썬의 float 타입은 Java의 double과 유사한 64비트 부동 소수점 숫자를 사용한다. 이는 약 15-17자리의 십진수 정밀도를 제공한다.
# 그러나 float 타입은 매우 큰 수나 매우 작은 수를 표현할 때 정밀도가 떨어질 수 있다. 이는 부동 소수점 연산의 한계 때문이다.
# 매우 큰 수나 매우 작은 수를 다루어야 하는 경우, decimal 모듈을 사용하여 고정 소수점 숫자를 사용할 수 있다. 
```

```python
v4 = True
print(type(v4))  # <class 'bool'>
```
