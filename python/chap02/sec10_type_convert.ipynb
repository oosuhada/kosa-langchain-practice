```python
str1 = "3.54"

float1 = float(str1)  # 3.54. 문자열을 실수로 변환한다. str1은 소수점이 포함된 숫자로 구성되어 있기 때문에 실수로 변환할 수 있다. float() 함수는 문자열이 실수 형태일 때 변환이 가능하다.
int1 = int(float1)  # 3. 실수를 정수로 변환한다. float1은 소수점이 포함된 숫자로 구성되어 있기 때문에 정수로 변환할 수 있다. int() 함수는 실수가 정수 형태로 변환될 수 있을 때 사용할 수 있다.
print(int1)  # 3. int1의 값을 출력한다. int1은 float1을 정수로 변환한 값이므로 3이 출력된다.

# 방법1 - 실수로 변환한 후 정수로 변환
int1 = int(float(str1)) 
print(int1)

# 방법2 - 반올림
int2 = round(float(str1))
print(int2)
```

```python
str2 = "3.14759"
float2 = round(float(str2), 2)  # 3.15. 문자열을 실수로 변환한 후, 소수점 둘째 자리까지 반올림한다. str2는 소수점이 포함된 숫자로 구성되어 있기 때문에 실수로 변환할 수 있다. round() 함수는 실수를 지정한 소수점 자리까지 반올림하는 데 사용된다. float(str2)는 str2를 실수로 변환한 값이고, round(float(str2), 2)는 그 값을 소수점 둘째 자리까지 반올림한 결과를 반환한다.
print(float2)  # 3.15. float2의 값을 출력한다. float2는 str2를 실수로 변환한 후 소수점 둘째 자리까지 반올림한 값이므로 3.15가 출력된다.
```

```python
import math

# 아래 str3 값을 정수 3으로 변환하시오

str3 = "2.536"
int3 = int(float(str3))  # 방법1 - 실수로 변환한 후 정수로 변환
print(int3)
int4 = round(float(str3))  # 방법2 - 반올림
print(int4)
int5 = int(float(str3) + 0.5)  # 방법3 - 반올림을 수학적으로 구현
print(int5)
int6 = math.ceil(float(str3))  # 방법4 - 올림
print(int6)
```

```python
print(bool("True"))  # True. 문자열 "True"는 빈 문자열이 아니므로 참으로 평가된다. bool() 함수는 문자열이 빈 문자열이 아닌 경우 True를 반환한다.
print(bool("False"))  # True. 문자열 "False"는 빈 문자열이 아니므로 참으로 평가된다. bool() 함수는 문자열이 빈 문자열이 아닌 경우 True를 반환한다.
print(eval("True"))  # True. eval() 함수는 문자열로 표현된 파이썬 표현식을 실행하여 그 결과를 반환한다. 문자열 "True"는 파이썬에서 참을 나타내는 예약어이므로 eval("True")는 True를 반환한다.
print(eval("False"))  # False. eval() 함수는 문자열로 표현된 파이썬 표현식을 실행하여 그 결과를 반환한다. 문자열 "False"는 파이썬에서 거짓을 나타내는 예약어이므로 eval("False")는 False를 반환한다.
print(bool(""))  # False. 빈 문자열은 거짓으로 평가된다. bool() 함수는 빈 문자열인 경우 False를 반환한다.
print(bool(0))  # False. 숫자 0은 거짓으로 평가된다. bool() 함수는 숫자 0인 경우 False를 반환한다.
print(bool(1))  # True. 숫자 1은 참으로 평가된다. bool() 함수는 숫자 0이 아닌 경우 True를 반환한다.
print(bool(-1))  # True. 숫자 -1은 참으로 평가된다. bool() 함수는 숫자 0이 아닌 경우 True를 반환한다.
print(bool([]))  # False. 빈 리스트는 거짓으로 평가된다. bool() 함수는 빈 리스트인 경우 False를 반환한다.
print(bool([0]))  # True. 리스트 [0]은 빈 리스트가 아니므로 참으로 평가된다. bool() 함수는 빈 리스트가 아닌 경우 True를 반환한다.
print(bool([1, 2, 3]))  # True. 리스트 [1, 2, 3]은 빈 리스트가 아니므로 참으로 평가된다. bool() 함수는 빈 리스트가 아닌 경우 True를 반환한다.
print(bool({}))  # False. 빈 딕셔너리는 거짓으로 평가된다. bool() 함수는 빈 딕셔너리인 경우 False를 반환한다.
print(bool(()))  # False. 빈 튜플은 거짓으로 평가된다. bool() 함수는 빈 튜플인 경우 False를 반환한다.
print(bool(None))  # False. None은 거짓으로 평가된다. bool() 함수는 None인 경우 False를 반환한다.
print(bool(" "))  # True. 문자열 " "는 빈 문자열이 아니므로 참으로 평가된다. bool() 함수는 문자열이 빈 문자열이 아닌 경우 True를 반환한다.

# if문에서 bool() 함수를 사용하면 조건문을 작성할 때 유용하게 사용할 수 있다.
var = []
if not var:  # var가 빈 리스트이므로 not var는 True가 된다. 따라서 if 블록이 실행된다.
    print("암시적 false")  # "암시적 false"가 출력된다. var가 빈 리스트이므로 not var는 True가 되고, if 블록이 실행되어 이 문장이 출력된다.
```
