```python
# 온도 센서에서 측정된 데이터를 JSON 문자열로 반환하는 함수
import json
import random
from datetime import datetime


def temperature_sensor():
    # 방식1
    # data = {}  # 빈 딕셔너리 생성
    # data["sensor"] = "온도"
    # data["value"] = random.randint(20, 30)  # 20에서 30 사이의 랜덤한 정수 생성
    # data["datetime"] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    # 방식2
    data = {
        "sensor": "온도",
        "value": random.randint(20, 30),
        "datetime": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    }

    # dict 자료형을 JSON 문자열로 변환하여 반환
    str_json = json.dumps(
        data, ensure_ascii=False
    )  # ensure_ascii=False는 한글이 깨지지 않도록 하는 옵션. ascii가 아니라 UTF-8로 인코딩한다는 의미이다.
    # dumps의 s는 복수형의 s가 아니라 string의 s이다. JSON 문자열로 변환한다는 의미이다.
    # json.dump()는 dict 자료형을 JSON 문자열로 변환하여 파일에 저장하는 함수이다. dump의 p는 put의 p이다. 파일에 저장한다는 의미이다.
    return str_json
```

```python
temperature_sensor()
```

```python
# 온도 데이터를 파일에 저장하는 함수
import time
import os


def save_temperature():
    file_path = os.path.join(os.getcwd(), "temperature_data.txt")
    with open(file_path, mode="at", encoding="UTF-8") as file:
        for i in range(5):
            data = temperature_sensor()
            print(data)  # 콘솔에 데이터 출력
            file.write(data + "\n")  # 파일에 데이터를 쓰고 줄 바꿈
            time.sleep(1)  # 1초 간격 주기적으로 출력


save_temperature();
```

```python
# 파일에서 온도 데이터를 읽어오는 함수
def read_temperature_from_file():
    file_path = os.path.join(os.getcwd(), "temperature_data.txt")
    with open(file_path, "rt", encoding="utf-8") as file:
        while True:
            line = file.readline()  # str_json
            if line == "":  # if not line:
                break
            # JSON 문자열을 dict 자료형으로 변환
            data = json.loads(line)
            print(data)


read_temperature_from_file()
```
