```python
scores: list[int] = [83, 90, 87, 90, 78] # 타입 힌트. 
print(type(scores)) # <class 'list'>
print(scores[0]) # 83
print(scores[2]) # 87
```

```python
for i in range(0, len(scores)):
    print(f"scores[{i}]: {scores[i]}")
```

```python
list1: list[int] = [3, 2, 6, 7]
list2: list[int] = []

for item in list1:
    if item % 2 == 0:
        list2.append(item**2)
print(list2) # [4, 36]
```

```python
list1: list[int] = [3, 2, 6, 7]
list2 = [ item**2 for item in list1 if item%2 == 0]
print(list2) # [4, 36]
```

```python
# list1과 list2의 요소를 묶어서 튜플로 만들고 싶다면 zip() 함수를 사용할 수 있다.
list1: list[int] = [1, 2, 3, 4, 5]
list2: list[int] = [10, 20, 30, 40, 50]
zip_object = zip(list1, list2)
# print(zip_object.__next__()) # (1, 10)
# print(zip_object.__next__()) # (2, 20)
# print(zip_object.__next__()) # (3, 30)
# print(zip_object.__next__()) # (4, 40)
# print(zip_object.__next__()) # (5, 50)
# print(zip_object.__next__()) # StopIteration 예외가 발생한다. zip 객체의 요소를 모두 꺼냈기 때문이다.

list_tuple = list(zip_object)
# print(list_tuple) # [] zip 객체의 요소를 모두 꺼냈기 때문에 빈 리스트가 나온다. zip 객체는 한 번 요소를 꺼내면 다시 사용할 수 없다.
print(list_tuple) # [(1, 10), (2, 20), (3, 30), (4, 40), (5, 50)] zip 객체의 요소를 모두 꺼내기 전에 리스트로 변환했기 때문에 정상적으로 튜플이 담긴 리스트가 나온다.
```

```python
numbers = [10, 20, 30]

# 요소 반복 방법1
for num in numbers:
    print(num)
    
# 요소 반복 방법2 (인덱스 활용)
for item in enumerate(numbers):
    print(f"{item[0]}: {item[1]}") # (0, 10), (1, 20), (2, 30) 튜플로 인덱스와 요소가 함께 나온다.
    
for index, item in enumerate(numbers):
    print(f"{index}: {item}") # (0, 10), (1, 20), (2, 30) 튜플로 인덱스와 요소가 함께 나온다.
    
# 범위를 이용하는 방법
for index in range(len(numbers)):
    print(numbers[index]) # 10, 20, 30 인덱스를 이용해서 요소에 접근한다.
    print(f"{index}: {numbers[index]}") # (0, 10), (1, 20), (2, 30) 인덱스와 요소가 함께 나온다.
```

```python
scores = [99, 95, 93, 91, 88, 84, 78, 75, 73, 70]

top3 = scores[:3] # 슬라이싱을 이용해서 상위 3개의 요소를 가져온다.
print(type(top3), top3) # [99, 95, 93]

bottom3 = scores[-3:] # 슬라이싱을 이용해서 하위 3개의 요소를 가져온다.
print(type(bottom3), bottom3) # [75, 73, 70]
```

```python
# 요소 패킹, 언패킹
member = ["파이썬", "서울시 강남구", "010-123-1234"]

name, address, telephone = member # 언패킹. 리스트의 요소를 각각의 변수에 할당한다.
print(name) # 파이썬
print(address) # 서울시 강남구
print(telephone) # 010-123-1234

name, *rest = member # 언패킹. 리스트의 첫 번째 요소는 name 변수에 할당되고 나머지 요소들은 rest 변수에 list 형태로 할당된다.
print(name) # 파이썬
print(rest) # ['서울시 강남구', '010-123-1234']
```

```python
# 요소 변경 - 인덱싱으로 변경시 변경값만 제공, 슬라이싱으로 변경시 변경값을 리스트로 제공해야 한다.
scores = [99, 95, 93, 91, 88, 84, 78, 75, 73, 70]
scores[-3:] = [0, 0 ,0] # 슬라이싱을 이용해서 하위 3개의 요소를 0으로 변경한다.
print(scores) # [99, 95, 93, 91, 88, 84, 78, 0, 0, 0]
scores[1] = [100] # 인덱스를 이용해서 요소를 100으로 변경한다. 리스트로 변경된다. #list → int 불일치 Pylance 경고
print(scores) # [99, [100], 93, 91, 88, 84, 78, 0, 0, 0]
scores[1] = 100 # 인덱스를 이용해서 요소를 100으로 변경한다. 정수로 변경된다.
print(scores) # [99, 100, 93, 91, 88, 84, 78, 0, 0, 0]
scores[1:2] = [90] # 슬라이싱을 이용해서 요소를 90으로 변경한다. 리스트로 변경된다.
print(scores) # [99, 90, 93, 91, 88, 84, 78, 0, 0, 0]
scores[1:2] = [80, 70] # 슬라이싱을 이용해서 요소를 80과 70으로 변경한다. 리스트로 변경된다.
print(scores) # [99, 80, 70, 91, 88, 84, 78, 0, 0, 0]
```

```python
scores = []

scores.append(100) # 리스트에 요소를 추가한다.
scores.append(90)
scores.append(70)
print(scores) # [100, 90, 70]

scores.insert(2, 80) # 리스트의 2번 인덱스에 요소를 추가한다.
print(scores) # [100, 90, 80, 70]
```

```python
# 슬라이싱으로 제거
scores = [99, 99, 99, 99, 95, 93, 91, 88, 84, 78, 75, 73, 70]

scores[-3:] = [] # 슬라이싱을 이용해서 하위 3개의 요소를 제거한다.
print(scores) # [99, 99, 99, 99, 95, 93, 91, 88, 84, 78]

scores[1:2] = [] # 슬라이싱을 이용해서 1번 인덱스의 요소를 제거한다.
print(scores) # [99, 99, 99, 95, 93, 91, 88, 84, 78]

del scores[0] # del 키워드/인덱싱을 이용해서 0번 인덱스의 요소를 제거한다.
print(scores) # [99, 99, 95, 93, 91, 88, 84, 78]

scores.remove(99) # remove() 메서드를 이용해서 99라는 값을 제거한다. 첫 번째로 나오는 99가 제거된다.
print(scores) # [99, 95, 93, 91, 88, 84, 78]

# 만약 요소값을 이용해서 모든 요소를 제거하고 싶다면? -> while문 사용
scores2 = [99, 99, 99, 99, 95, 93, 91, 88, 84, 78]
while 99 in scores2:
    scores2.remove(99)
print(scores2) # [95, 93, 91, 88, 84, 78]
# 주의할 점. 99를 삭제하지 않는다면 99 in scores2 때문에 무한반복이 일어난다

# 모든 요소를 제거하는 방법
scores2.clear()
print(scores2)

# 리스트의 요소를 제거하는 방법은 여러 가지가 있다. 인덱스를 이용해서 제거하는 방법, 슬라이싱을 이용해서 제거하는 방법, remove() 메서드를 이용해서 제거하는 방법 등이 있다. 상황에 따라 적절한 방법을 선택해서 사용하면 된다.
```
