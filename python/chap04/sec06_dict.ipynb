#### - 딕셔너리 생성

```python
from typing import Any

member: dict[str, Any] = {
    "id": "python",
    "name": "파이썬",
    "age": 20,
    "tel": "0101231234",
}
```

#### - 요소 찾기

```python
name = member["name"]
print(name)
print(member)

# job = member["job"]  # 키가 없으면 KeyError 예외가 발생한다.
job = member.get(
    "job", "직업 정보 없음"
)  # get() 메서드는 키가 없을 때 기본값을 반환한다.
print(job)
```

```python
member = {
    "name": "홍길동",
    "grade": 3,
    "adult": False,
    "hobby": ["영화", "축구"],
    "school": ("대치초등학교", "서초중학교", "강남고등학교"),
    "family": {"father": "홍삼원", "mother": "김순희"},
}

# 멤버가 졸업한 중학교
middle_school = member["school"][1]
print(middle_school)

# 멤버의 엄마의 이름
mother_name = member["family"]["mother"]
print(mother_name)
```

```python
for key in member:
    value = member[key]
    print(f"{key}: {value}")
```

```python
member.keys()  # 키 목록을 반환한다.
member.values()  # 값 목록을 반환한다.
member.items()  # 키-값 쌍의 목록을 반환한다.
for key, value in member.items():
    print(f"{key}: {value}")
print()
for key in member.keys():
    print(key)
print()
for value in member.values():
    print(value)
```

```python
for item in member.items():
    print(type(item), item)
print()
for key, value in member.items():
    print(type(key), key, type(value), value)
```

#### - 요소 변경

```python
board = {
    "title": "파이썬 도서",
    "writer": "신용권",
    "date": "20180731",
    "content": "AI 엔지니어를 위한 파이썬 언어",
    "hitcount": 10000,
}

board["title"] = "파이썬 프로그래밍"
board["hitcount"] += 1

print(board)
```

#### - 요소 추가

```python
# board에서 add, append 기능을 사용할 수 없다. 동적으로 요소를 추가하려면 키를 지정해서 값을 할당해야 한다.
board = {
    "title": "파이썬 도서",
    "writer": "신용권",
    "date": "20180731",
    "content": "AI 엔지니어를 위한 파이썬 언어",
    "hitcount": 10000,
}

board["publisher"] = "한빛미디어"  # 새로운 키-값 쌍을 추가한다.
print(board)
```

#### - 요소 제거

```python
board = {
    "title": "파이썬 도서",
    "writer": "신용권",
    "date": "20180731",
    "content": "AI 엔지니어를 위한 파이썬 언어",
    "hitcount": 10000,
}

del board["date"]  # 키-값 쌍을 제거한다.
del board["hitcount"]
print(board)
```
