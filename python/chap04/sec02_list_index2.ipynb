```python
scores = [92, 95, 76, 100, 88, 87, 100]

# 홀수만 따로 모아서 list로 생성하는 방법
odd_scores = [ score for score in scores if score % 2 == 1]
print(odd_scores)
```

```python
# 요소 정렬
scores = [92, 95, 76, 100, 88, 87, 100]
scores.sort()
print(scores) # [76, 87, 88, 92, 95, 100, 100]

scores.sort(reverse=True)
print(scores) # [100, 100, 95, 92, 88, 87, 76]
```

```python
# 리스트 복사

list1 = [1, 2, 3]

# 참조 (같은 객체)
list2 = list1
list2.append(99)
print(list1)  # [1, 2, 3, 99] ← 같이 바뀜!
print(list2)  # [1, 2, 3, 99]

# 얕은 복사 (새 객체)
list1 = [1, 2, 3]
list3 = list1.copy()
list3.append(99)
print(list1)  # [1, 2, 3] ← 그대로
print(list3)  # [1, 2, 3, 99]

# 깊은 복사 (중첩 리스트)
import copy
list4 = [[1, 2], [3, 4]]
list5 = list4.copy()      # 얕은 복사
list6 = copy.deepcopy(list4)  # 깊은 복사

list4[0].append(99)
print(list5)  # [[1, 2, 99], [3, 4]] ← 내부까지 같이 바뀜
print(list6)  # [[1, 2], [3, 4]]     ← 완전히 독립
```

```python
# 리스트 확장
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.extend(list2)
print(list1) # [1, 2, 3, 4, 5, 6]
```

```python
# 리스트 연산
original = [1, 2, 3]

list1 = original
list2 = [4, 5, 6]

list1 = list1 + list2
print(list1) # [1, 2, 3, 4, 5, 6]
print(original) # [1, 2, 3] original은 변경되지 않음
# + 리스트 연산자는 새 리스트를 만들어서 list1에 재할당하기 때문에 original이 변경되지 않음
```

```python
# 전체의 평균 구하기
class1 = [10, 20, 30]
class2 = [40, 50, 60, 70, 80]

# 리스트 결합 후 평균 구하기
all_scores = class1 + class2
average = sum(all_scores) / len(all_scores)
print(average) # 45.0

# 2중 for 문으로 평균 구하기
total = 0
count = 0
print([class1, class2]) # [[10, 20, 30], [40, 50, 60, 70, 80]]
for class_scores in [class1, class2]:
    for score in class_scores:
        total += score
        count += 1
average = total / count
print(average) # 45.0   

scores = [[10, 20, 30],[40, 50, 60, 70, 80]]
total: int = 0
count: int = 0
for i in range(len(scores)):
    for j in range(len(scores[i])):
        total += scores[i][j]
        count += 1
average = total / count
print(average) # 45.0   
```
