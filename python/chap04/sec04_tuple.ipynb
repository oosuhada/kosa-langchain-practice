```python
# 튜플의 경우 요소 수와 값을 변경할 수 없다. (불변‼️)

grade = (1, 2, 3, 4, 5)
season = ("봄", "여름", "가을", "겨울")
day = ("월", "화", "수", "목", "금", "토")

print(grade, f"요소수: {len(grade)}")
print(season, f"요소수: {len(season)}")
print(day, f"요소수: {len(day)}")
```

```python
# 튜플은 괄호가 없이도 요소를 나열하면 튜플로 인식한다.
color = "빨강", "노랑", "파랑"
print(color, f"요소수: {len(color)}")
```

```python
# 튜플의 타입 힌트

# ...은 요소의 개수가 정해지지 않았음을 의미한다.
grade: tuple[int, ...] = (
    1,
    2,
    3,
    4,
    5,
)
season: tuple[str, ...] = ("봄", "여름", "가을", "겨울")

print(grade, f"요소수: {len(grade)}")
print(season, f"요소수: {len(season)}")

person: tuple[str, int] = (
    "홍길동",
    30,
)  # 요소의 개수가 2개로 고정되고, 첫 번째 요소는 문자열, 두 번째 요소는 정수임을 의미한다.
print(person, f"요소수: {len(person)}")
```

```python
# 튜플의 언패킹

person: tuple[str, int] = ("홍길동", 30)
name, age = person  # 튜플의 요소를 각각 변수에 할당하는 것을 언패킹이라고 한다.
print(f"이름: {name}, 나이: {age}")
```

```python
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

# 튜플의 덧셈과 곱셈
tuple3 = tuple1 + tuple2  # 튜플의 덧셈은 두 튜플을 이어붙이는 것이다.
tuple4 = tuple1 * 3  # 튜플의 곱셈은튜플을 여러 번 이어붙이는 것이다.
print(tuple3)
print(tuple4)
```

```python
names = ("홍길동", "김철수", "이영희")

for index, value in enumerate(names):
    print(f"{index}: {value}")
```
