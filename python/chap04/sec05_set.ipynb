#### - set 생성

```python
# 값의 목록으로 set 생성
nums = {1, 2, 3, 4, 5, 3}
print(nums)

# list로 set 생성
names = ["홍길동", "김철수", "이영희", "홍길동"]
names = set(names)
print(type(names), names)

# 빈 {}로는 set을 만들 수 없다. {}는 빈 dict를 만든다. (set()으로 빈 set을 만들어야 한다.)
# cities = {}
# print(type(cities), cities) #dict 타입이 나옴

# empty_set = {} --- IGNORE ---
empty_set = set()
print(type(empty_set), empty_set)
```

#### - 요소 얻기

```python
chars = {"a", "b", "c", "d", "e"}

# set은 index로 관리가 되지 않기 때문에 원하는 요소를 뽑을 수 없다. (인덱싱, 슬라이싱 불가)

while chars:
    print(
        chars.pop()
    )  # set에서 요소를 하나씩 꺼내는 pop() 메서드. set은 순서가 없기 때문에 어떤 요소가 꺼내질지는 예측할 수 없다.

chars2 = {"a", "b", "c", "d", "e"}

while chars2:
    value = chars2.pop()
    if value == "d" or value == "e":
        print(value)
```

#### - 요소 추가

```python
pocket = set()

pocket.add("구슬1")
pocket.add("구슬2")
print(pocket)

pocket.update(["구슬3", "구슬4"])
print(pocket)
```

#### - 요소 제거

```python
basket = {"사과", "바나나", "오렌지", "수박", "포도"}

basket.remove(
    "수박"
)  # remove() 메서드는 제거하려는 요소가 없으면 KeyError 예외가 발생한다.
print(basket)
```

#### - set 연산

```python
set1 = {"a", "b", "c", "d", "e"}
set2 = {"d", "e", "f", "g", "h"}
set3 = {"e", "f", "g", "h", "i"}
set4 = {"f", "g", "h", "i", "j"}

# 합집합
result = set1.union(set2)
print(result)

result = set1 | set2
print(result)

# 교집합
result = set1.intersection(set2)
print(result)

result = set1 & set2
print(result)

# 차집합
result = set1.difference(set2)
print(result)
result = set1 - set2
print(result)
```
