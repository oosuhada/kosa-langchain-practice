#### - Stack 구조

```python
stack = []

# 요소 넣기
stack.append(100)
stack.append(50)
stack.append(500)

# 요소 가져오기 1
print(stack.pop())
print(stack.pop())
print(stack.pop())
# print(stack.pop()) # 빈 리스트에서 pop()을 호출하면 IndexError가 발생
print(stack)  # []

stack.append(100)
stack.append(50)
stack.append(500)

# 요소 가져오기 2
for i in range(len(stack)):
    print(stack.pop())

stack.append(100)
stack.append(50)
stack.append(500)
print(stack)  # []

# 요소 가져오기 3
while len(stack) > 0:
    print(stack.pop())
print(stack)  # []
```

#### - Queue

```python
from collections import deque

# 큐 객체 생성
queue = deque()  # FIFO(First In First Out) 방식

# 메시지 저장하기
queue.append({"from": "Alice", "to": "Bob", "message": "Hello, Bob!"})
queue.append({"from": "Bob", "to": "Alice", "message": "Hi, Alice!"})
queue.append({"from": "Alice", "to": "Bob", "message": "How are you?"})
queue.append({"from": "Bob", "to": "Alice", "message": "I'm good, thanks!"})

# 메시지 처리하기 1
# 큐에서 요소를 가져올 때는 popleft() 메서드를 사용하여 가장 먼저 들어온 요소를 가져옴
# pop을 사용하면 가장 나중에 들어온 요소를 먼저 가져오게 됨
print(queue.popleft())
print(queue.popleft())
print(queue.popleft())
print(queue.popleft())
print(queue)  # deque([])

# 메시지 처리하기 2
while queue:
    print(queue.popleft())
print(queue)  # deque([])
```
