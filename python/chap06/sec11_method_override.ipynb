```python
from typing import override


class Parent:
    def method(self):
        print("Parent method called : 100줄")


class Child(Parent):
    @override
    def method(self):
        super().method()  # 부모 클래스의 method() 호출  - 100줄 재사용
        print("Child method called : 50줄")


c = Child()
c.method()  # Child method called
```
