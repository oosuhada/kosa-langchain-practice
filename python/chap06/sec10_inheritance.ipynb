```python
class Person:
    def __init__(self, name: str) -> None:
        self.name = name


p1 = Person("Gabriel")
print
```

```python
class Student(Person):
    def __init__(self, name: str, sno: str) -> None:
        super().__init__(name)
        self.sno = sno


s1 = Student("Gabriel", "123456789")
print(s1.name)
print(s1.sno)
```

```python
class A:
    def method1(self):
        print("A.method1() 실행")


class B:
    def method2(self):
        print("B.method2() 실행")


class C:
    def method3(self):
        print("C.method3() 실행")


# 다중 상속 - java와 다르게 python에서는 다중 상속이 가능하다.
class D(A, B, C):
    pass


d = D()
d.method1()  # A.method1() 실행
d.method2()  # B.method2() 실행
d.method3()  # C.method3() 실행


# 상속 순서
class E(C, A, B):
    pass


e = E()
e.method3()  # C.method3() 실행
e.method2()  # A.method2() 실행
e.method1()  # B.method1() 실행


# 상속 순서 변경
class F(A, C, B):
    pass


f = F()
f.method1()  # A.method1() 실행
f.method3()  # C.method3() 실행
f.method2()  # B.method2() 실행
```
