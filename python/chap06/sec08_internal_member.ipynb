```python
class Member:
    def __init__(self, name: str, ssn: str) -> None:
        self.__name = name
        self.__ssn = ssn
        # 이중 언더바(___) : 강한 접근제한을 하겠다는 의미로 사용. Java의 private 메소드를 흉내내기 위해 사용했음.

    def get_name(self) -> str:
        return self.__name

    def get_ssn(self) -> str:
        return self.__ssn
```

```python
member = Member("홍길동", "123456-1234567")
# print(member.__name) # AttributeError
# print(member.__ssn) # AttributeError
print(
    member._Member__name
)  # 홍길동 # 이중 언더바도 Name mangling을 통해서 접근할 수 있다. 완전히 접근 차단은 힘들다.
print(member._Member__ssn)  # 123456-1234567

# 이상적인 방법. get method를 사용해서 그러나 Java에서 많이 쓰는 방식이지 파이썬스럽지는 않다. -> 그래서 Sec09 파일.
print(member.get_name())  # 홍길동
print(member.get_ssn())  # 123456-1234567
```
