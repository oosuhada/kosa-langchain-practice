#### - 클래스 정의


```python
class Car:
    pass


car1 = Car()  # 내부적으로는 __new__() 메소드가 자동으로 호출되어 인스턴스(객체) 생성,
# __new__() 메소드로 생성된 인스턴스(객체)를 __init__()으로 초기화한다.
car2 = Car()
# car1, car2는 각자 다른 객체 인스턴스(객체)를 생성한다.

# 객체 번지수 조회
print(id(car1))  # 인스턴스 생성 후 id 값을 출력한다.
print(id(car2))  # 인스턴스 생성 후 id 값을 출력한다.
```

```python
# car1 = Car()이 실행될때 파이썬 내부적으로 흐름

# 1. Car.__new__(Car) 호출
# 2. __new__가 새 인스턴스(객체)를 생성해서 반환
# 3. 반환된 객체가 Car 타입이면 Car.__init__(self) 호출
# 4. 초기화가 끝난 객체가 car1에 할당
```
