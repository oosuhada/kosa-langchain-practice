```python
from typing import override


class Car:
    def __init__(self, model: str, color: str) -> None:
        self.model = model
        self.color = color

    @override
    def __str__(self) -> str:
        # print(my_car), str(my_car)를 호출할 때 사용된다.
        # 사람이 읽기 쉬운 출력용 문자열을 반환하는 메소드다.
        # __str__이 없으면 __repr__이 대신 사용될 수 있다.
        return f"__str__ method called : Car(model={self.model}, color={self.color})"

    @override
    def __repr__(self) -> str:
        # repr(my_car)를 호출하거나, 리스트/딕셔너리 안에서 객체가 출력될 때 주로 사용된다.
        # __str__이 없으면 print(obj), str(obj)에서도 대신 사용될 수 있다.
        # 개발자가 디버깅할 때 객체 상태를 명확히 확인하기 위한 문자열을 반환하는 메소드다.
        return f"__repr__ method called : Car(model={self.model}, color={self.color})"


my_car = Car("Sonata", "White")

print(my_car)  # __str__ method called : Car(model=Sonata, color=White)
print(my_car.__repr__())  # __repr__ method called : Car(model=Sonata, color=White)
print(my_car.__str__())  # __str__ method called : Car(model=Sonata, color=White)

# print(
#     my_car
# )  # <__main__.Car object at 0x000001F9C0D0F0B0> - 메모리 진수. 우리에게는 쓸모없는 정보
# print(my_car.__repr__())  # <__main__.Car object at 0x000001F9C0D0F0B0>
# print(my_car.__str__())  # <__main__.Car object at 0x000001F9C0D0F0B0>

print(
    repr(my_car)
)  # __repr__ method called : Car(model=Sonata, color=White) # repr() 내장함수를 사용하면 __repr__ 메소드가 호출된다.
print(str(my_car))  # str() 내장함수를 사용하면 __str__ 메소드가 호출된다.
my_car  # __str__ method called : Car(model=Sonata, color=White)
```

```python
# __str__과 __repr__의 사용 우선순위
#
# 1. print(obj)
#    - obj.__str__()을 먼저 사용한다.
#    - __str__이 정의되어 있지 않으면 __repr__()을 대신 사용한다.
#
# 2. str(obj)
#    - obj.__str__()을 먼저 사용한다.
#    - __str__이 정의되어 있지 않으면 __repr__()을 대신 사용한다.
#
# 3. repr(obj)
#    - obj.__repr__()을 사용한다.
#
# 4. 리스트/튜플/딕셔너리 같은 컨테이너 안에서 객체를 출력할 때
#    - 각 객체의 __repr__()을 사용한다.
#
# 정리:
# - __str__만 있으면 print(obj), str(obj)는 __str__을 사용한다.
# - __repr__만 있으면 print(obj), str(obj), repr(obj)가 모두 __repr__을 사용한다.
# - 둘 다 있으면 print(obj), str(obj)는 __str__을 사용하고,
#   repr(obj)와 컨테이너 내부 출력은 __repr__을 사용한다.
```
