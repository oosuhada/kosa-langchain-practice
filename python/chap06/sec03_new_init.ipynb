```python
from typing import override, Self


class Car:
    """Car 클래스 정의"""

    @override  # 부모 클래스(object)의 메소드를 재정의한다는 의미
    def __new__(cls) -> Self:
        """객체를 생성하는 특수 메소드 -> Java의 클래스 메소드(정적 메소드) new
        객체 생성 전에 클래스(cls)를 기준으로 호출되는 특수 메소드
        객체가 아직 없기 때문에 self가 아니라 cls를 첫 번째 인자로 받는다."""
        print("Car.__new__() 메소드 실행")
        return super().__new__(cls)
        # 만약 return cls()를 사용하면 다시 Car()를 호출하게 되어 __new__가 반복 호출될 수 있다. (에러 발생)

    @override
    def __init__(self) -> None:
        """생성된 객체를 초기화하는 인스턴스 메소드
        self는 __new__가 만든 인스턴스 자기 자신이다."""
        print("Car.__init__() 메소드 실행")
        super().__init__()


car1 = Car()
# 매개값 없이 Car.__new__() 메소드 실행
# __new__()로 객체를 만들고, 이어서 __init__()으로 초기화한다.
```

```python
# 같은 이름의 Car 클래스를 다시 정의하면, 이전 Car 클래스는 이 이름으로 접근할 수 없게 된다.
class Car:
    # 객체(생성자) 생성 메소드
    # 잘 정의하지 않는다고 했는데 언제 정의해야 하는가? str, int 값이 정확하게 들어왔는지 검사하는 코드를 넣기 위해서 사용
    # __new__는 일반적으로 자주 정의하지 않는다. 객체 생성 자체를 제어해야 하는 특별한 경우에 사용할 수 있다.
    # 유효성 검사에 실패하면 아예 객체를 만들지 말아야 한다.면... 파이썬은 유효성 검사가 없이는 str, int 값이 아니어도 실행은 정상적으로 동작해버린다.
    # 일반적인 값 검사는 물론 __init__에서도 가능하다. 객체 생성 자체를 막아야 하는 특별한 경우 __new__에서 검사할 수 있다.
    # 잘못된 타입을 막고 싶다면 isinstance() 같은 런타임 검사를 직접 작성해야 한다.
    # 일반적인 값 검사는 __init__에서 처리해도 된지만 잘못된 객체 생성 자체를 막아야 하는 특별한 경우에는 __new__에서 검사할 수 있다.
    # new에서 검사하고 올바른 객체만 생성해야지 init에서 유효성 검사를 실행하면 안된다
    def __new__(cls, model: str, year: int) -> Self:
        # isinstance(value, type)은 value가 해당 타입의 객체인지 확인한다.
        # 맞으면 True, 아니면 False를 반환한다.
        if not isinstance(model, str):
            raise TypeError("model은 str 타입이어야 합니다.")
        if not isinstance(year, int):
            raise TypeError("year은 int 타입이어야 합니다.")
        # # 매개변수가 유효할 경우에만 부모 object의 __new__를 호출해 실제 객체(인스턴스)를 생성한다.
        return super().__new__(
            cls
        )  # (심화) 싱글톤으로 만들고 싶을때 이 부분은 반복 호출되지 않도록 해야한다

    # 객체(생성자) 초기화 메소드
    # 앞으로 수업에서 생성자라고 하면 보통 이 __init__을 의미한다.
    def __init__(self, model: str, year: int) -> None:
        # self는 __new__가 생성한 인스턴스 자기 자신이다.
        self.model = model
        self.year = year

    def __str__(self) -> str:
        return f"{self.model} {self.year}"


# model, year 2개의 값을 전달해서 Car 인스턴스를 생성한다.
car1 = Car("Honda", 2020)  # 생성 시 인스턴스 변수 model, year를 초기화 한다.
car2 = Car("Toyota", 2021)
# car3 = Car(2020, "Honda")  # __new__의 isinstance 검사에서 TypeError 발생

print(car1)  # Honda 2020
print(car2)  # Toyota 2021
```

#### - init(): 객체의 초기화를 담당하는 매직 메소드


```python
class Car:
    # @override - 이 경우에는 오히려 override를 붙이면 안된다,.
    def __init__(
        self, name: str, price: int, year: str = "2026"
    ) -> None:  # 항상 return값은 None를 반환해야 한다.
        # 객체를 받아야 하므로 정의할때는 self값이 반드시 있어야 한다.
        # 생성자를 호출할때는 self값이 없어도 됨.

        # 인스턴스 필드 선언과 초기화
        self.name = name
        self.price = price
        self.year = year  # default value가 있기 때문에 맨 뒤에 선언했음
```

```python
my_car = Car("G80", 8000, "2026")
# 인스턴스 필드에 접근하기 - 값을 읽는다
print(my_car.name)  # G80
print(my_car.price)  # 8000
print(my_car.year)  # 2026

my_car = Car("G80", 8000)
print(my_car.name)  # G80
print(my_car.price)  # 8000
print(my_car.year)  # 2026
```

```python
class Member:
    def __init__(
        self, name: str, age: int = 25, *hobbies, **other
    ) -> None:  # 가변길이 *, ** 에는 타입힌트를 주지 않는다.
        print(type(hobbies))  # tuple
        print(type(other))  # dict
        self.name = name
        self.age = age
        self.hobbies = hobbies
        self.other = other

    def __str__(self) -> str:
        return f"{self.name} {self.age}"
```

```python
m1 = Member("장우수", 34, "영화", "음악", "게임", tel="010-1234-5678", city="서울")
print(m1.name)  # 장우수
print(m1.age)  # 34
print(m1.hobbies)  # (영화, 음악, 게임)
print(m1.hobbies[0])  # 영화
print(m1.other)  # {"tel": "010-1234-5678", "city": "서울"}
print(m1.other["tel"])  # 010-1234-5678
print(m1.other.get("tel"))  # 010-1234-5678
```

```python
m1 = Member("장우수", 34, tel="010-1234-5678", city="서울")
print(m1.name)  # 장우수
print(m1.age)  # 34
print(m1.hobbies)
print(m1.other["tel"])  # 010-1234-5678
```
