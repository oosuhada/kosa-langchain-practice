```python
class Car:
    def __init__(self, speed: int = 0, tire: str | None = None) -> None:
        # tire는 문자열이거나 None일 수 있으므로 str | None으로 타입 힌트를 작성한다.
        # int도 None을 허용하려면 int | None처럼 명시해야 한다.
        self.speed = speed
        self.tire = tire

    # 클래스 안에 정의된 함수는 인스턴스 메소드가 된다.
    # 인스턴스 메소드의 첫 번째 매개변수 self는 호출한 객체 자기 자신이다.
    def run(self) -> int:
        # 같은 객체의 다른 인스턴스 메소드를 호출할 때도 self를 사용한다.
        self.set_speed(60)  # Java의 this.speed와 비슷하게 Python에서는 self.speed를 사용한다.
        print(f"Car is running at {self.speed} km/h")
        return self.speed

    def set_speed(self, speed: int) -> None:
        self.speed = speed

    def change_tire(self, tire: str) -> str:
        self.tire = tire
        print(f"Car tire is changed to {self.tire}")
        return self.tire

    def stop(self) -> None:
        self.speed = 0
        print("Car is stopped")


car1 = Car(speed=100, tire="기본 타이어")
car1.run()
car1.stop()


# 클래스 밖에 정의된 함수는 별개의 함수다. 자동으로 인스턴스 메소드가 되지 않는다.
# self라는 이름도 클래스 밖에서는 특별한 의미가 아니라 일반 매개변수 이름일 뿐이다.
# def run(self) -> None:
#     self.speed = 60
#     print(f"Car is running at {self.speed} km/h")
```

```python
my_car = Car()

current_speed = my_car.run()  # run() 내부에서 speed를 60으로 바꾸고 60을 반환한다.
my_car.set_speed(100)  # 객체의 현재 speed 필드는 100으로 변경된다.
print(my_car.speed)  # 100
print(current_speed)  # 60
```

#### - 동적 메소드 바인딩

```python
# 파이썬에서는 실행 중에 특정 인스턴스에만 메소드를 추가할 수도 있다.
# 다만 클래스에 메소드를 미리 정의하는 방식이 더 일반적이고 읽기 쉽다.
# 동적 메소드 바인딩은 가능하다는 점을 이해하는 정도로 정리한다.

import types

my_car = Car()


# 클래스 밖에 정의된 일반 함수다.
# 첫 번째 매개변수 self는 이름만 self일 뿐, 아직 my_car와 자동 연결되어 있지 않다.
def dynamic_stop(self) -> None:
    self.speed = 0
    print("Dynamic stop")


# types.MethodType(function, instance)을 사용하면
# 일반 함수를 특정 인스턴스에 바인딩된 메소드로 만들 수 있다.
my_car.dynamic_stop = types.MethodType(dynamic_stop, my_car)

my_car.set_speed(100)
print(my_car.speed)  # 100
my_car.dynamic_stop()
print(my_car.speed)  # 0

# dynamic_stop은 my_car 인스턴스에만 추가된 메소드다.
# 다른 Car 인스턴스에는 자동으로 생기지 않는다.
other_car = Car()
# other_car.dynamic_stop()  # AttributeError: 'Car' object has no attribute 'dynamic_stop'
```
