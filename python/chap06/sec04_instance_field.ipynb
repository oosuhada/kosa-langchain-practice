```python
class Member:
    def __init__(self, name: str, age: int, hobbies: list[str] | None = None) -> None:
        # __init__에서 self.name, self.age, self.hobbies를 만들면
        # 이 객체는 name, age, hobbies라는 인스턴스 필드를 가지게 된다.
        self.name = name
        self.age = age
        # 기본값으로 []를 직접 쓰면 여러 인스턴스가 같은 리스트를 공유할 수 있다.
        # 그래서 None을 기본값으로 두고, 함수 안에서 새 리스트를 만들어 주는 방식이 안전하다.
        self.hobbies = hobbies if hobbies is not None else []
```

```python
m1 = Member("장우수", 34, ["영화", "음악"])

# 기존 인스턴스 필드 값 변경
m1.name = "장소영"
m1.age = 31
m1.hobbies[1] = "피아노"

# hobbies 필드가 가리키는 리스트 자체를 새 리스트로 교체
m1.hobbies = ["게임", "투자"]
m1.hobbies.append("다이빙")

print(m1.name)
print(m1.age)
print(m1.hobbies)
```

#### - 동적 필드 지원
