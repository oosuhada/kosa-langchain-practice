```python
class Car:
    # 클래스 필드
    kind = "승용차"

    @classmethod
    def method1(
        cls,
    ):  # @classmethod가 없이는 pylance가 Argument missing for parameter "cls" 경고
        print("method1() 실행")
        print(Car.kind)  # 승용차
        print(cls.kind)  # 승용차
        # Car로 들어오는 클래스나 cls로 들어오는 클래스나 같은 클래스기 때문에 같다.

    @staticmethod
    def method2():  # @staticmethod 없이는 Cannot access attribute "method2" for class "type[Car]"
        print("method2() 실행")
        print(Car.kind)  # 승용차
        # print(cls.kind)  cls를 사용하지 않았기 때문에 cls.kind는 사용할 수 없다

    def method3(self):
        print("method3() 실행")


Car.method1()
Car.method2()
# Car.method3() # def method3(self): self가 들어가면 instance method이므로 Car.method3() 방식으로 실행할 수 없다.
```

#### - 싱글톤 구현하기


```python
# Java Spring 방식 : 자동으로 싱글톤이 됨. 어노테이션 하나만 적용.
# @Service
# class MemberService {}


from typing import Self


class MemberService:
    # 클래스 변수
    _member_service = None
    _initialized = False

    # 객체 생성 역할
    def __new__(cls) -> Self:
        if cls._member_service == None:
            # 객체를 최초로 생성할 경우에는 super().__new__(cls)를 사용한다.
            # 잘 이해했는지 모르겠다. 변수가 없어도 생성할 수 있는데 그래서 클래스 변수를 사용한다고 하신거 같다.
            cls._member_service = super().__new__(cls)
        return cls._member_service

    # 객체 초기화 역할 = default 값으로 세팅하겠다는 의미
    def __init__(self) -> None:
        # if self._member_service != None:
        if self._initialized == False:
            print("__init__() 메소드 실행")
            self.field1 = ""
            self.field2 = ""
            self._initialized = True

    # 이대로 실행하면 __init__() 메소드가 두번 실행된다. 싱글톤이 아니게 된다. 그래서 if self.member_service == None 조건 이 아니라 _initialized = False 값 을 추가해서 해결한다. 생성자 호출때마다 new, init이 반복 호출되는 것을 방지하는 것이 목적이다.

    # 클래스 메소드를 이용해서 싱글톤 객체 반환
    @classmethod
    def get_instance(cls) -> Self:
        return cls()
```

```python
member_service_1 = MemberService()
member_service_2 = MemberService()

# print(member_service_1 == member_service_2)  # 동등 비교
print(member_service_1 is member_service_2)  # 번지 비교
```

```python
member_service_3 = MemberService.get_instance()
member_service_4 = MemberService.get_instance()
print(member_service_3 is member_service_4)  # True
```
