```python
class Member:
    # 초기화 메소드
    def __init__(self, name: str, ssn: str) -> None:
        self.__name: str = name
        # ssn은 직접 __ssn에 넣지 않고 property setter를 거치게 한다.
        # 그래야 객체를 처음 만들 때도 주민번호 형식 검사가 실행된다.
        self.ssn = ssn

    # 읽기 속성 정의
    @property
    def name(self) -> str:
        return self.__name

    # 쓰기 속성 정의
    @name.setter
    def name(self, name: str) -> None:
        self.__name = name

    # 읽기 속성 정의
    @property
    def ssn(self) -> str:
        return self.__ssn

    # 쓰기 속성 정의
    @ssn.setter
    def ssn(self, ssn: str) -> None:
        # 주민번호 전체 14자리 양식: 숫자 6자리 + 하이픈(-) + 숫자 7자리
        # 예: 123456-1234567
        if len(ssn) != 14:
            raise ValueError("주민번호는 하이픈을 포함해 전체 14자리여야 합니다.")
        if ssn[6] != "-":
            raise ValueError("주민번호 7번째 문자는 하이픈(-)이어야 합니다.")

        front = ssn[:6]
        back = ssn[7:]
        if not front.isdigit() or not back.isdigit():
            raise ValueError("주민번호는 하이픈을 제외하고 숫자로만 구성되어야 합니다.")

        self.__ssn = ssn


member = Member("홍길동", "123456-1234567")
# 읽기 속성 사용
print(
    member.name
)  # def name(self) 메소드를 실행하여 self.__name 값을 반환한다. -> 홍길동
# 쓰기 속성 사용
member.name = "이중길동"  # 새로 저장할 값
print(member.name)  # 이중길동
print(member._Member__name)  # 이중길동


member.ssn = "123456-1234567"
# 읽기 속성 사용
print(member.ssn)  # 123456-1234567
print(member._Member__ssn)  # 123456-1234567

# SSN에 대해서 주민번호 전체 14자리 양식이 다르면 raise ValueError 예외를 발생시키도록 속성을 정의해보시오
# 아래 예제들은 모두 setter에서 ValueError가 발생해야 한다.
invalid_ssn = [
    "1234561234567",  # 하이픈 없음
    "12345-1234567",  # 앞자리 길이 부족
    "123456-123456",  # 뒷자리 길이 부족
    "123456/1234567",  # 하이픈 위치 문자 오류
    "12345a-1234567",  # 숫자가 아닌 문자 포함
]

for invalid_ssn in invalid_ssn:
    try:
        member.ssn = invalid_ssn
    except ValueError as e:
        print(f"{invalid_ssn}: {e}")
```
