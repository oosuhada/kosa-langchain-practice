```python
from dataclasses import dataclass


# __init__(), __str__(), __repr__(), __eq__() 등의 메서드를 자동으로 생성해주는 데코레이터
@dataclass
class Member:
    mid: str
    mname: str
    mage: int


m1 = Member("abc123", "홍길동", 30)
print(m1)
m1  # 내부의 __repr__() 메서드가 호출되어 객체의 문자열 표현이 출력된다.

m2 = Member("abc123", "홍길동", 30)

print(m1 == m2)  # 동등 비교: __eq__() 메서드가 호출되어 객체의 내용이 같은지 비교한다.
print(m1 is m2)  # 번지 비교: m1과 m2는 서로 다른 객체이므로 False가 출력된다.

m3 = Member("abc123", "홍길동", 31)

print(
    m1 == m3
)  # 동등 비교: __eq__() 메서드가 호출되어 객체의 내용이 다른지 비교한다. age가 다르므로 False가 출력된다.
print(m1 is m3)  # 번지 비교: m1과 m3는 서로 다른 객체이므로 False가 출력된다.

m4 = Member("abc456", "김성수", 40)
print(m4)

# 이름이 다르면 아래와 같은 예외가 발생한다.
# TypeError: cannot set attribute
# m4.mname = "김성수"

# 이름이 다르면 아래와 같은 예외가 발생한다.
# TypeError: cannot set attribute
# m4.mage = 40
m4
```
