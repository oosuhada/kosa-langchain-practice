```python
available_coffee = 5
while available_coffee > 0: # False 값이 될 때까지 반복
    print("커피를 한 잔 드릴까요?")
    available_coffee -= 1
    print(f"남은 커피의 수: {available_coffee}잔")
print("커피가 다 떨어졌습니다. 죄송합니다.")
```

```python
stop: bool = False
while not stop: # stop이 False인 동안 반복
    print("커피를 한 잔 드릴까요?")
    stop = True # stop을 True로 설정하여 반복 종료
print("커피가 다 떨어졌습니다. 죄송합니다.")
```

```python
print("a", end="") # end 매개변수는 print 함수가 출력한 후에 어떤 문자열을 출력할지를 지정하는 매개변수이다. 기본값은 "\n"으로, 줄바꿈을 의미한다. end=""로 설정하면 줄바꿈 대신 빈 문자열이 출력되어, 다음 print 함수의 출력이 같은 줄에 이어서 나타나게 된다.
print("b", end="\n") # end="\n"은 print 함수가 출력한 후에 줄바꿈을 하도록 지정하는 것이다. 기본값이므로 생략해도 동일한 결과를 얻을 수 있다.
print("c", end="")
```

```python
stop: bool = False

while not stop: # stop이 False인 동안 반복
    print("입력: ", end="")
    keyboard_input = input() # 사용자로부터 입력을 받는다. input() 함수는 사용자가 입력한 값을 문자열로 반환한다.
    if keyboard_input == "음악":
        print("음악을 재생합니다.")
    elif keyboard_input == "영화":
        print("영화를 재생합니다.")
    elif keyboard_input == "중지": # 사용자가 "중지"를 입력하면 stop을 True로 설정하여 반복을 종료한다.
        print("재생을 중지합니다.")
        stop = True
print("프로그램을 종료합니다.")
    
```
