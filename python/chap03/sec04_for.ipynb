```python
# 대괄호로 리스트 생성
scores: list[int] = [90, 86, 75, 92, 95]

# 총합을 저장하는 변수
score_sum: int = 0

# list 항목을 반복해서 가져와서 누적하기
for score in scores:  # iterable(반복 가능한) 객체가 들어와야 한다. 시퀀스 타입도 iterable이지만, for문은 시퀀스 타입만이 아니라 iterable이면 모두 사용 가능하다. scores는 list 타입이므로 sequence 타입 객체이다.
    score_sum += score

# 평균 계산
score_avg: float = score_sum / len(scores)
print("평균 점수:", score_avg)

# 시퀀스 타입 - 리스트(list), 튜플(tuple), 범위(range), 문자열(str)
# 반복 가능한 타입(iterable) - 대표적으로 dict, map, set 등이 있다. 시퀀스 타입도 반복 가능한 타입이지만, 반복 가능한 타입이 시퀀스 타입은 아니다. 반복 가능한 타입은 시퀀스 타입이 아니더라도 for문에서 사용할 수 있다. 예를 들어, dict는 반복 가능한 타입이지만 시퀀스 타입이 아니다. dict의 경우, for문에서 키(key)를 반복한다.
```

```python
member: dict[str, int] = {"김철수": 30, "이영희": 25, "박민수": 28}
member2: dict = {"id": "python", "name": "김철수", "age": 30}

# for문을 이용해서 dict 객체값을 가져올 수 있다.
# dict 안에 key값이 3개이므로 for문이 3번 반복된다. key값이 "김철수", "이영희", "박민수"인 순서로 반복된다.
# (dict는 순서가 없는 자료형이므로, key값의 순서는 실행할 때마다 달라질 수 있다.)
# Python 3.7부터 dict는 삽입 순서를 보장한다. 따라서 key값의 순서는 항상 삽입한 순서대로 반복된다.
for key in member2:
    value = member2[key]
    print(key, ":", value)
```

```python
for i in range(10):
    print(i)

for i2 in range(1, 11):
    print(i2)

# range는 함수일까 생성자일까?
# range는 함수가 아니라 클래스(class)다.
# # range(...)는 range 클래스의 객체를 생성한다. 함수처럼 호출하지만 결과는 반복 가능한 range 객체다.
# range(10)은 range 클래스의 인스턴스(객체)를 생성하는 것이며,
# 생성된 range 객체는 시퀀스 타입이므로 for문에서 iterable로 사용할 수 있다.
```

```python
# 구구단을 작성하는 코드를 작성해보세요 예를 들면 2 * 1 = 2, 2 * 2 = 4, ... 9 * 9 = 81
for x in range(2, 10):
    for y in range(1, 10):
        print(f"{x} * {y} = {x * y}")
    print()  # 단이 바뀔 때마다 개행을 해서 구구단이 구분되도록 한다.
```
