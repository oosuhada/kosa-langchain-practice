```python
# 1~10 범위를 반복할때 5가 되면 반복을 중지
# while문에서 break문을 사용할 것
i = 1
while i <= 10:
    if i == 6:
        break
    print(i)
    i += 1
```

```python
# 1~10 범위를 반복할때 5가 되면 반복을 중지
# for문에서 break문을 사용할 것
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

```python
# 1~10 범위를 반복하면서 3의 배수의 합을 출력하세요
# for문에서 continue문을 사용할 것
total = 0
for i in range(1, 11):
    if i % 3 != 0:
        continue
    total += i
print(total)
```

```python
# 1~10 범위를 반복하면서 3의 배수의 합을 출력하세요
# for문에서 continue문을 사용할 것
total = 0
for i in range(1, 11):
    if i % 3 == 0:
        total += i
print(total)
```
