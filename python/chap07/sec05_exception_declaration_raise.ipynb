```python
class BankError(Exception):  # 상위 예외 클래스인 Exception을 상속받음
    def __init__(self, message: str = "", error_code: str = "") -> None:
        super().__init__(
            message
        )  # 상위 클래스의 __init__ 메서드를 호출하여 message를 전달
        self.message = message  # message를 인스턴스 변수로 저장
        self.error_code = error_code  # error_code를 인스턴스 변수로 저장


# 예외 발생 코드 예시
# raise BankError()
# raise BankError("계좌가 존재하지 않습니다.") # BankError 예외를 발생시키고 메시지를 전달
# raise BankError("잔액이 부족합니다.") # BankError 예외를 발생시키고 메시지를 전달


class Bank:
    def __init__(self) -> None:
        self.balance: int = 0

    def withdraw(self, amount: int) -> None:
        if amount > self.balance:
            raise BankError(
                "잔액이 부족합니다.", "E001"
            )  # BankError 예외를 발생시키고 메시지와 에러 코드를 전달
        self.balance -= amount

    def deposit(self, amount: int) -> None:
        self.balance += amount

    # def transfer(self, amount: int, recipient: Bank) -> None:
    #     if amount > self.balance:
    #         raise BankError("잔액이 부족합니다.")
    #     recipient.deposit(amount)
    #     self.withdraw(amount)


try:
    bank = Bank()

    bank.balance = 10000
    bank.withdraw(5000)
    print(bank.balance)

    bank.withdraw(
        6000
    )  # BankError 예외가 발생하여 "잔액이 부족합니다." 메시지가 출력됨
except (
    BankError
) as e:  # 여기서 예외가 발생한다면 참조하는 변수 e를 사용해서 메시지를 출력함
    message = str(e)  # 예외 객체 e의 메시지를 출력
    error_code = e.error_code  # error_code를 인스턴스 변수로 저장
    if error_code == "E001":  # 에러 코드가 E001인 경우
        print(f"[{error_code}] {message}")
    # 출력 결과: [E001] 잔액이 부족합니다.
    elif error_code == "E002":  # 에러 코드가 E002인 경우
        print(f"[{error_code}] {message}")
    # 출력 결과: [E002] 계좌가 존재하지 않습니다.
    elif error_code == "E003":  # 에러 코드가 E003인 경우
        print(f"[{error_code}] {message}")
    # 출력 결과: [E003] 송금할 계좌가 존재하지 않습니다.
    elif error_code == "E004":  # 에러 코드가 E004인 경우
        print(f"[{error_code}] {message}")
    # 출력 결과: [E004] 비밀번호가 틀립니다.
```
