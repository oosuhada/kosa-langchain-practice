```python
while True:
    try:
        print("숫자를 입력하세요")
        x = input()
        result = (
            10 / int(x)
        )  # 만약 입력된 값이 str이나 0이라면 ValueError나 ZeroDivisionError가 발생할 것이다.
    except:  # java에서 catch(Exception e)와 같은 역할을 하는 except문은 모든 예외를 잡아내는 역할을 한다. 하지만 어떤 예외가 발생했는지 알 수 없기 때문에, except문을 여러 개 작성하여 각각의 예외에 대해서 다른 메시지를 출력하도록 하는 것이 좋다.
        print("0 또는 숫자가 아닌 값을 입력하셨습니다. 다시 입력해주세요.")
    else:
        print("연산 결과: ", result)
    finally:
        print("계속 여부 (y/n)?")
        exit = input()
        if exit == "n":
            break
```
