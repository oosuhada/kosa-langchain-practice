```python
while True:
    try:
        print("숫자를 입력하세요")
        x = input()
        result = (
            10 / int(x)
        )  # 만약 입력된 값이 str이나 0이라면 ValueError나 ZeroDivisionError가 발생할 것이다.
    except ZeroDivisionError:
        print("0이 아닌 다른 숫자를 입력해야 합니다. 다시 입력해주세요.")
    except ValueError:
        print("숫자가 아닌 값을 입력하셨습니다. 다시 입력해주세요.")
    except Exception:  # ZeroDivisionError와 ValueError의 상위 예외 클래스기 때문에 ZeroDivisionError와 ValueError보다 먼저 작성하면 모든 예외처리를 Exception에서 처리하기 때문에 ZeroDivisionError와 ValueError는 작성할 수 없다.
        print("알 수 없는 오류가 발생했습니다. 다시 입력해주세요.")
    else:
        print("연산 결과: ", result)
    finally:
        print("계속 여부 (y/n)?")
        exit = input()
        if exit == "n":
            break
```
