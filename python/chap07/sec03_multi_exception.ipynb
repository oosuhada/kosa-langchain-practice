```python
while True:
    try:
        print("숫자를 입력하세요")
        x = input()
        result = (
            10 / int(x)
        )  # 만약 입력된 값이 str이나 0이라면 ValueError나 ZeroDivisionError가 발생할 것이다.
    except ZeroDivisionError:
        print("0이 아닌 다른 숫자를 입력해야 합니다. 다시 입력해주세요.")
    except ValueError:
        print("숫자가 아닌 값을 입력하셨습니다. 다시 입력해주세요.")
    else:
        print("연산 결과: ", result)
    finally:
        print("계속 여부 (y/n)?")
        exit = input()
        if exit == "n":
            break
```
